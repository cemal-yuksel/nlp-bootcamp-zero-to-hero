## 1.3 🏗️ GPT-2 Mimarisi

<div style="background-color: #fff3e0; padding: 15px; border-left: 5px solid #ff9800; border-radius: 5px;">
    <h3 style="color: #e65100; margin-top: 0;">🔑 GPT-2 = Generative Pre-trained Transformer 2</h3>
    <p><strong>OpenAI</strong> tarafından geliştirilmiş <strong>decoder-only</strong> transformer modelidir.</p>
</div>

### � Temel Özellikler Tablosu:

| Özellik | Değer | Açıklama |
|---------|-------|----------|
| **📦 Parametre Sayısı** | 124M → 1.5B | Small, medium, large, XL varyantları |
| **🏛️ Mimari** | Transformer Decoder | Sadece decoder katmanları kullanır |
| **🔤 Vocab Size** | 50,257 | Farklı token sayısı |
| **📏 Max Uzunluk** | 1024 token | ~768 kelime |
| **📚 Eğitim Verisi** | 40GB WebText | 8M web sayfası |
| **🎯 Görev** | Causal LM | Sıradaki tokeni tahmin et |
| **⚡ Attention Heads** | 12 (small) → 25 (XL) | Multi-head attention |
| **🔢 Hidden Size** | 768 (small) → 1600 (XL) | Embedding boyutu |

### 🔄 GPT-2 İşlem Akışı:

<div style="background-color: #f5f5f5; padding: 20px; border-radius: 10px; font-family: monospace; font-size: 14px;">

```
INPUT: "The movie was" 
   ↓
🔡 [TOKENIZATION] 
   → Token IDs: [464, 3807, 373]
   ↓
📊 [EMBEDDING LAYER] 
   → Her token 768 boyutlu vektöre dönüşür
   ↓
📍 [POSITIONAL ENCODING] 
   → Kelime sırası bilgisi eklenir
   ↓
🔄 [12x TRANSFORMER BLOCKS]
   ├─ 🎯 Multi-Head Self-Attention (12 head)
   │  └─ Her kelime diğer kelimelerle ilişki kurar
   ├─ 🔧 Layer Normalization
   ├─ 🧮 Feed-Forward Network (3072 nöron)
   └─ ➕ Residual Connections
   ↓
📤 [OUTPUT LAYER] 
   → 50,257 boyutlu olasılık dağılımı
   ↓
✅ PREDICTION: "amazing" (en yüksek olasılık: 0.32)
```

</div>

> **⚡ Hız Avantajı:** Transformer, RNN/LSTM'in aksine tüm tokenleri **paralel** işler. Bu, eğitim süresini 10-100x azaltır!

## 1.5 🔑 Temel Kavramlar

### 1️⃣ 🔤 Tokenization

<div style="background-color: #e8f5e9; padding: 15px; border-left: 5px solid #4caf50; border-radius: 5px;">
    <p><strong>Tokenization:</strong> Metni model için anlaşılır parçalara (token) ayırma işlemidir.</p>
</div>

**📌 Örnek:**
```python
Metin: "ChatGPT is amazing!"

🔽 Tokenization

Tokenler: ["Chat", "G", "PT", " is", " amazing", "!"]
Token IDs: [30813, 38, 11571, 318, 4998, 0]
```

> **💡 İpucu:** GPT-2, **Byte-Pair Encoding (BPE)** kullanır. Sık kullanılan kelimeler tek token, nadir kelimeler birden fazla token olur!

---

### 2️⃣ 🎯 Attention Mekanizması

<div style="background-color: #fff3e0; padding: 15px; border-left: 5px solid #ff9800; border-radius: 5px;">
    <p><strong>Attention:</strong> Modelin, bir kelimenin diğer kelimelerle olan ilişkisini öğrenmesidir.</p>
</div>

**📌 Örnek Cümle:** *"The cat sat on the mat"*

```
"sat" kelimesi için attention ağırlıkları:

🐱 "cat"  ████████████ 0.85  ← Yüksek (Özne - kim oturdu?)
🪑 "sat"  ██████ 0.45         ← Orta (Kendisi)
📍 "on"   ███ 0.20            ← Orta (Konum bilgisi)
🟦 "mat"  ██████████ 0.75     ← Yüksek (Nesne - nerede?)
📰 "the"  █ 0.05              ← Düşük (Yardımcı kelime)
```

> **🔥 Self-Attention Gücü:** Model, "sat" kelimesini anlamak için "cat" (özne) ve "mat" (nesne) kelimelerine yüksek attention verir!

### 3️⃣ 🌡️ Temperature

<div style="background-color: #e3f2fd; padding: 15px; border-left: 5px solid #2196F3; border-radius: 5px;">
    <p><strong>Temperature:</strong> Üretim <strong>çeşitliliğini</strong> ve <strong>yaratıcılığını</strong> kontrol eden hiperparametredir.</p>
</div>

| Temperature | Davranış | Kullanım Alanı | Örnek |
|-------------|----------|----------------|-------|
| **0.0 - 0.3** | 🎯 Deterministik | Teknik dokümantasyon, kod | "The capital of France is **Paris**." |
| **0.5 - 0.7** | ⚖️ Dengeli | Genel metin üretimi | "The capital of France is **Paris, a beautiful city**." |
| **0.8 - 1.0** | 🎨 Yaratıcı | Blog yazıları, hikayeler | "The capital of France is **Paris, where art dances with history**." |
| **1.5+** | 🎲 Çok yaratıcı | Deneysel, sanatsal | "The capital of France is **Paris, though some say it's moonlight**." |

**📊 Görsel Açıklama:**
```
Temperature = 0.1          Temperature = 1.0
┌─────────────┐           ┌─────────────┐
│ Paris: 95%  │███████    │ Paris: 40%  │████
│ London: 3%  │▓          │ London: 25% │███
│ Berlin: 1%  │           │ Rome: 20%   │██
│ Rome: 1%    │           │ Madrid: 15% │██
└─────────────┘           └─────────────┘
   Güvenli                   Çeşitli
```

### 4️⃣ 🎲 Top-k ve Top-p Sampling

<div style="background: linear-gradient(to right, #fa709a 0%, #fee140 100%); padding: 15px; border-radius: 10px; color: white;">
    <h4 style="color: white; margin-top: 0;">Sampling Stratejileri</h4>
    <p>Model, bir sonraki tokeni seçerken kullandığı yöntemlerdir.</p>
</div>

#### 🔸 **Top-k Sampling**

En yüksek **k** olasılıklı tokenden birini seç.

```python
top_k = 50 → En olası 50 token arasından seç

Örnek:
Token Olasılıkları:
1. "amazing" → 0.25  ✅ Seçilebilir (Top 50'de)
2. "great"   → 0.20  ✅ Seçilebilir
...
50. "okay"   → 0.001 ✅ Seçilebilir
51. "weird"  → 0.0009 ❌ Seçilemez (50'den sonra)
```

#### 🔸 **Top-p (Nucleus) Sampling**

Kümülatif olasılığı **p**'ye ulaşan tokenleri seç.

```python
top_p = 0.9 → Kümülatif %90'a ulaşan token kümesinden seç

Örnek:
Token          Olasılık    Kümülatif
1. "amazing"   0.35        0.35  ✅
2. "great"     0.25        0.60  ✅
3. "good"      0.20        0.80  ✅
4. "nice"      0.12        0.92  ❌ (0.9 geçildi)
5. "okay"      0.08        1.00  ❌
```

**📊 Karşılaştırma:**

| Strateji | Avantaj | Dezavantaj | Ne Zaman Kullan? |
|----------|---------|------------|------------------|
| **Greedy** | Deterministik, hızlı | Tekrarlayıcı | Kod üretimi |
| **Top-k** | Kontrollü çeşitlilik | Sabit k her zaman uygun olmaz | Genel kullanım |
| **Top-p** | Dinamik, esnek | Hesaplama maliyeti | Yaratıcı metin |
| **Beam Search** | En iyi global seçim | Çok yavaş | Çeviri, özetleme |

### 2️⃣ 🎯 Attention Mekanizması

<div style="background-color: #fff3e0; padding: 15px; border-left: 5px solid #ff9800; border-radius: 5px;">
    <p><strong>Attention:</strong> Modelin, bir kelimenin diğer kelimelerle olan ilişkisini öğrenmesidir.</p>
</div>

**📌 Örnek Cümle:** *"The cat sat on the mat"*

```
"sat" kelimesi için attention ağırlıkları:

🐱 "cat"  ████████████ 0.85  ← Yüksek (Özne - kim oturdu?)
🪑 "sat"  ██████ 0.45         ← Orta (Kendisi)
📍 "on"   ███ 0.20            ← Orta (Konum bilgisi)
🟦 "mat"  ██████████ 0.75     ← Yüksek (Nesne - nerede?)
📰 "the"  █ 0.05              ← Düşük (Yardımcı kelime)
```

> **🔥 Self-Attention Gücü:** Model, "sat" kelimesini anlamak için "cat" (özne) ve "mat" (nesne) kelimelerine yüksek attention verir!

### 4️⃣ 🎲 Top-k ve Top-p Sampling

<div style="background: linear-gradient(to right, #fa709a 0%, #fee140 100%); padding: 15px; border-radius: 10px; color: white;">
    <h4 style="color: white; margin-top: 0;">Sampling Stratejileri</h4>
    <p>Model, bir sonraki tokeni seçerken kullandığı yöntemlerdir.</p>
</div>

#### 🔸 **Top-k Sampling**

En yüksek **k** olasılıklı tokenden birini seç.

```python
top_k = 50 → En olası 50 token arasından seç

Örnek:
Token Olasılıkları:
1. "amazing" → 0.25  ✅ Seçilebilir (Top 50'de)
2. "great"   → 0.20  ✅ Seçilebilir
...
50. "okay"   → 0.001 ✅ Seçilebilir
51. "weird"  → 0.0009 ❌ Seçilemez (50'den sonra)
```

#### 🔸 **Top-p (Nucleus) Sampling**

Kümülatif olasılığı **p**'ye ulaşan tokenleri seç.

```python
top_p = 0.9 → Kümülatif %90'a ulaşan token kümesinden seç

Örnek:
Token          Olasılık    Kümülatif
1. "amazing"   0.35        0.35  ✅
2. "great"     0.25        0.60  ✅
3. "good"      0.20        0.80  ✅
4. "nice"      0.12        0.92  ❌ (0.9 geçildi)
5. "okay"      0.08        1.00  ❌
```

**📊 Karşılaştırma:**

| Strateji | Avantaj | Dezavantaj | Ne Zaman Kullan? |
|----------|---------|------------|------------------|
| **Greedy** | Deterministik, hızlı | Tekrarlayıcı | Kod üretimi |
| **Top-k** | Kontrollü çeşitlilik | Sabit k her zaman uygun olmaz | Genel kullanım |
| **Top-p** | Dinamik, esnek | Hesaplama maliyeti | Yaratıcı metin |
| **Beam Search** | En iyi global seçim | Çok yavaş | Çeviri, özetleme |

### 🔧 Yardımcı Fonksiyonlar

Bu notebook boyunca kullanacağımız özel fonksiyonları tanımlayalım.

````python
def print_section(title, emoji="📌"):
    """
    Bölüm başlıklarını görsel olarak yazdıran fonksiyon
    
    Args:
        title (str): Başlık metni
        emoji (str): Başlangıç emojisi
    """
    print("\n" + "="*70)
    print(f"{emoji} {title}")
    print("="*70 + "\n")


def visualize_attention_weights(text, attention_scores, tokenizer):
    """
    Attention ağırlıklarını ısı haritası olarak görselleştirir
    
    Args:
        text (str): Girdi metni
        attention_scores (tensor): Attention skorları
        tokenizer: Tokenizer nesnesi
    """
    tokens = tokenizer.tokenize(text)
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(attention_scores.cpu().numpy(), 
                xticklabels=tokens, 
                yticklabels=tokens,
                cmap='YlOrRd', 
                annot=False,
                cbar_kws={'label': 'Attention Ağırlığı'})
    plt.title('Self-Attention Ağırlıkları', fontsize=14, pad=20)
    plt.xlabel('Key Tokens', fontsize=12)
    plt.ylabel('Query Tokens', fontsize=12)
    plt.tight_layout()
    plt.show()


def compare_generations(prompt, texts_dict, max_display=150):
    """
    Farklı parametrelerle üretilen metinleri karşılaştırır
    
    Args:
        prompt (str): Başlangıç promptu
        texts_dict (dict): {parametre_adı: üretilen_metin} dictionary
        max_display (int): Gösterilecek maksimum karakter sayısı
    """
    print_section("🔍 Metin Üretimi Karşılaştırması", "📊")
    print(f"💬 Prompt: '{prompt}'\n")
    
    for i, (param_name, text) in enumerate(texts_dict.items(), 1):
        # Promptu kaldır
        generated_part = text[len(prompt):].strip()
        display_text = generated_part[:max_display]
        
        if len(generated_part) > max_display:
            display_text += "..."
        
        print(f"{i}. {param_name}")
        print(f"   └─ {display_text}")
        print(f"   📏 Uzunluk: {len(generated_part)} karakter\n")


def calculate_perplexity(model, tokenizer, text, device):
    """
    Bir metin için perplexity (şaşkınlık) değerini hesaplar
    Düşük perplexity = Model metni daha iyi anlıyor
    
    Args:
        model: Dil modeli
        tokenizer: Tokenizer
        text (str): Değerlendirilecek metin
        device: torch.device
    
    Returns:
        float: Perplexity değeri
    """
    encodings = tokenizer(text, return_tensors='pt')
    input_ids = encodings.input_ids.to(device)
    
    with torch.no_grad():
        outputs = model(input_ids, labels=input_ids)
        loss = outputs.loss
        perplexity = torch.exp(loss)
    
    return perplexity.item()


def plot_token_probabilities(probabilities, tokens, top_k=10):
    """
    En yüksek olasılıklı tokenleri görselleştirir
    
    Args:
        probabilities (tensor): Token olasılıkları
        tokens (list): Token listesi
        top_k (int): Gösterilecek top token sayısı
    """
    top_probs, top_indices = torch.topk(probabilities, top_k)
    top_tokens = [tokens[idx] for idx in top_indices]
    
    plt.figure(figsize=(10, 6))
    plt.barh(range(top_k), top_probs.cpu().numpy(), color='steelblue')
    plt.yticks(range(top_k), top_tokens)
    plt.xlabel('Olasılık', fontsize=12)
    plt.title(f'En Olası {top_k} Token', fontsize=14, pad=20)
    plt.gca().invert_yaxis()
    
    # Değerleri çubukların üzerine yaz
    for i, prob in enumerate(top_probs.cpu().numpy()):
        plt.text(prob + 0.001, i, f'{prob:.4f}', va='center')
    
    plt.tight_layout()
    plt.show()


print("✅ Yardımcı fonksiyonlar tanımlandı!")
print("📋 Fonksiyonlar:")
print("   • print_section() - Başlık yazdırma")
print("   • visualize_attention_weights() - Attention görselleştirme")
print("   • compare_generations() - Metin karşılaştırma")
print("   • calculate_perplexity() - Perplexity hesaplama")
print("   • plot_token_probabilities() - Token olasılıkları grafiği")
````

### 2️⃣ 🎯 Attention Mekanizması

<div style="background-color: #fff3e0; padding: 15px; border-left: 5px solid #ff9800; border-radius: 5px;">
    <p><strong>Attention:</strong> Modelin, bir kelimenin diğer kelimelerle olan ilişkisini öğrenmesidir.</p>
</div>

**📌 Örnek Cümle:** *"The cat sat on the mat"*

```
"sat" kelimesi için attention ağırlıkları:

🐱 "cat"  ████████████ 0.85  ← Yüksek (Özne - kim oturdu?)
🪑 "sat"  ██████ 0.45         ← Orta (Kendisi)
📍 "on"   ███ 0.20            ← Orta (Konum bilgisi)
🟦 "mat"  ██████████ 0.75     ← Yüksek (Nesne - nerede?)
📰 "the"  █ 0.05              ← Düşük (Yardımcı kelime)
```

> **🔥 Self-Attention Gücü:** Model, "sat" kelimesini anlamak için "cat" (özne) ve "mat" (nesne) kelimelerine yüksek attention verir!

### 4️⃣ 🎲 Top-k ve Top-p Sampling

<div style="background: linear-gradient(to right, #fa709a 0%, #fee140 100%); padding: 15px; border-radius: 10px; color: white;">
    <h4 style="color: white; margin-top: 0;">Sampling Stratejileri</h4>
    <p>Model, bir sonraki tokeni seçerken kullandığı yöntemlerdir.</p>
</div>

#### 🔸 **Top-k Sampling**

En yüksek **k** olasılıklı tokenden birini seç.

```python
top_k = 50 → En olası 50 token arasından seç

Örnek:
Token Olasılıkları:
1. "amazing" → 0.25  ✅ Seçilebilir (Top 50'de)
2. "great"   → 0.20  ✅ Seçilebilir
...
50. "okay"   → 0.001 ✅ Seçilebilir
51. "weird"  → 0.0009 ❌ Seçilemez (50'den sonra)
```

#### 🔸 **Top-p (Nucleus) Sampling**

Kümülatif olasılığı **p**'ye ulaşan tokenleri seç.

```python
top_p = 0.9 → Kümülatif %90'a ulaşan token kümesinden seç

Örnek:
Token          Olasılık    Kümülatif
1. "amazing"   0.35        0.35  ✅
2. "great"     0.25        0.60  ✅
3. "good"      0.20        0.80  ✅
4. "nice"      0.12        0.92  ❌ (0.9 geçildi)
5. "okay"      0.08        1.00  ❌
```

**📊 Karşılaştırma:**

| Strateji | Avantaj | Dezavantaj | Ne Zaman Kullan? |
|----------|---------|------------|------------------|
| **Greedy** | Deterministik, hızlı | Tekrarlayıcı | Kod üretimi |
| **Top-k** | Kontrollü çeşitlilik | Sabit k her zaman uygun olmaz | Genel kullanım |
| **Top-p** | Dinamik, esnek | Hesaplama maliyeti | Yaratıcı metin |
| **Beam Search** | En iyi global seçim | Çok yavaş | Çeviri, özetleme |

### 🔧 Yardımcı Fonksiyonlar

Bu notebook boyunca kullanacağımız özel fonksiyonları tanımlayalım.

````python
def print_section(title, emoji="📌"):
    """
    Bölüm başlıklarını görsel olarak yazdıran fonksiyon
    
    Args:
        title (str): Başlık metni
        emoji (str): Başlangıç emojisi
    """
    print("\n" + "="*70)
    print(f"{emoji} {title}")
    print("="*70 + "\n")


def visualize_attention_weights(text, attention_scores, tokenizer):
    """
    Attention ağırlıklarını ısı haritası olarak görselleştirir
    
    Args:
        text (str): Girdi metni
        attention_scores (tensor): Attention skorları
        tokenizer: Tokenizer nesnesi
    """
    tokens = tokenizer.tokenize(text)
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(attention_scores.cpu().numpy(), 
                xticklabels=tokens, 
                yticklabels=tokens,
                cmap='YlOrRd', 
                annot=False,
                cbar_kws={'label': 'Attention Ağırlığı'})
    plt.title('Self-Attention Ağırlıkları', fontsize=14, pad=20)
    plt.xlabel('Key Tokens', fontsize=12)
    plt.ylabel('Query Tokens', fontsize=12)
    plt.tight_layout()
    plt.show()


def compare_generations(prompt, texts_dict, max_display=150):
    """
    Farklı parametrelerle üretilen metinleri karşılaştırır
    
    Args:
        prompt (str): Başlangıç promptu
        texts_dict (dict): {parametre_adı: üretilen_metin} dictionary
        max_display (int): Gösterilecek maksimum karakter sayısı
    """
    print_section("🔍 Metin Üretimi Karşılaştırması", "📊")
    print(f"💬 Prompt: '{prompt}'\n")
    
    for i, (param_name, text) in enumerate(texts_dict.items(), 1):
        # Promptu kaldır
        generated_part = text[len(prompt):].strip()
        display_text = generated_part[:max_display]
        
        if len(generated_part) > max_display:
            display_text += "..."
        
        print(f"{i}. {param_name}")
        print(f"   └─ {display_text}")
        print(f"   📏 Uzunluk: {len(generated_part)} karakter\n")


def calculate_perplexity(model, tokenizer, text, device):
    """
    Bir metin için perplexity (şaşkınlık) değerini hesaplar
    Düşük perplexity = Model metni daha iyi anlıyor
    
    Args:
        model: Dil modeli
        tokenizer: Tokenizer
        text (str): Değerlendirilecek metin
        device: torch.device
    
    Returns:
        float: Perplexity değeri
    """
    encodings = tokenizer(text, return_tensors='pt')
    input_ids = encodings.input_ids.to(device)
    
    with torch.no_grad():
        outputs = model(input_ids, labels=input_ids)
        loss = outputs.loss
        perplexity = torch.exp(loss)
    
    return perplexity.item()


def plot_token_probabilities(probabilities, tokens, top_k=10):
    """
    En yüksek olasılıklı tokenleri görselleştirir
    
    Args:
        probabilities (tensor): Token olasılıkları
        tokens (list): Token listesi
        top_k (int): Gösterilecek top token sayısı
    """
    top_probs, top_indices = torch.topk(probabilities, top_k)
    top_tokens = [tokens[idx] for idx in top_indices]
    
    plt.figure(figsize=(10, 6))
    plt.barh(range(top_k), top_probs.cpu().numpy(), color='steelblue')
    plt.yticks(range(top_k), top_tokens)
    plt.xlabel('Olasılık', fontsize=12)
    plt.title(f'En Olası {top_k} Token', fontsize=14, pad=20)
    plt.gca().invert_yaxis()
    
    # Değerleri çubukların üzerine yaz
    for i, prob in enumerate(top_probs.cpu().numpy()):
        plt.text(prob + 0.001, i, f'{prob:.4f}', va='center')
    
    plt.tight_layout()
    plt.show()


print("✅ Yardımcı fonksiyonlar tanımlandı!")
print("📋 Fonksiyonlar:")
print("   • print_section() - Başlık yazdırma")
print("   • visualize_attention_weights() - Attention görselleştirme")
print("   • compare_generations() - Metin karşılaştırma")
print("   • calculate_perplexity() - Perplexity hesaplama")
print("   • plot_token_probabilities() - Token olasılıkları grafiği")
````

### 2️⃣ 🎯 Attention Mekanizması

<div style="background-color: #fff3e0; padding: 15px; border-left: 5px solid #ff9800; border-radius: 5px;">
    <p><strong>Attention:</strong> Modelin, bir kelimenin diğer kelimelerle olan ilişkisini öğrenmesidir.</p>
</div>

**📌 Örnek Cümle:** *"The cat sat on the mat"*

```
"sat" kelimesi için attention ağırlıkları:

🐱 "cat"  ████████████ 0.85  ← Yüksek (Özne - kim oturdu?)
🪑 "sat"  ██████ 0.45         ← Orta (Kendisi)
📍 "on"   ███ 0.20            ← Orta (Konum bilgisi)
🟦 "mat"  ██████████ 0.75     ← Yüksek (Nesne - nerede?)
📰 "the"  █ 0.05              ← Düşük (Yardımcı kelime)
```

> **🔥 Self-Attention Gücü:** Model, "sat" kelimesini anlamak için "cat" (özne) ve "mat" (nesne) kelimelerine yüksek attention verir!

### 4️⃣ 🎲 Top-k ve Top-p Sampling

<div style="background: linear-gradient(to right, #fa709a 0%, #fee140 100%); padding: 15px; border-radius: 10px; color: white;">
    <h4 style="color: white; margin-top: 0;">Sampling Stratejileri</h4>
    <p>Model, bir sonraki tokeni seçerken kullandığı yöntemlerdir.</p>
</div>

#### 🔸 **Top-k Sampling**

En yüksek **k** olasılıklı tokenden birini seç.

```python
top_k = 50 → En olası 50 token arasından seç

Örnek:
Token Olasılıkları:
1. "amazing" → 0.25  ✅ Seçilebilir (Top 50'de)
2. "great"   → 0.20  ✅ Seçilebilir
...
50. "okay"   → 0.001 ✅ Seçilebilir
51. "weird"  → 0.0009 ❌ Seçilemez (50'den sonra)
```

#### 🔸 **Top-p (Nucleus) Sampling**

Kümülatif olasılığı **p**'ye ulaşan tokenleri seç.

```python
top_p = 0.9 → Kümülatif %90'a ulaşan token kümesinden seç

Örnek:
Token          Olasılık    Kümülatif
1. "amazing"   0.35        0.35  ✅
2. "great"     0.25        0.60  ✅
3. "good"      0.20        0.80  ✅
4. "nice"      0.12        0.92  ❌ (0.9 geçildi)
5. "okay"      0.08        1.00  ❌
```

**📊 Karşılaştırma:**

| Strateji | Avantaj | Dezavantaj | Ne Zaman Kullan? |
|----------|---------|------------|------------------|
| **Greedy** | Deterministik, hızlı | Tekrarlayıcı | Kod üretimi |
| **Top-k** | Kontrollü çeşitlilik | Sabit k her zaman uygun olmaz | Genel kullanım |
| **Top-p** | Dinamik, esnek | Hesaplama maliyeti | Yaratıcı metin |
| **Beam Search** | En iyi global seçim | Çok yavaş | Çeviri, özetleme |

### 🔧 Yardımcı Fonksiyonlar

Bu notebook boyunca kullanacağımız özel fonksiyonları tanımlayalım.

````python
def print_section(title, emoji="📌"):
    """
    Bölüm başlıklarını görsel olarak yazdıran fonksiyon
    
    Args:
        title (str): Başlık metni
        emoji (str): Başlangıç emojisi
    """
    print("\n" + "="*70)
    print(f"{emoji} {title}")
    print("="*70 + "\n")


def visualize_attention_weights(text, attention_scores, tokenizer):
    """
    Attention ağırlıklarını ısı haritası olarak görselleştirir
    
    Args:
        text (str): Girdi metni
        attention_scores (tensor): Attention skorları
        tokenizer: Tokenizer nesnesi
    """
    tokens = tokenizer.tokenize(text)
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(attention_scores.cpu().numpy(), 
                xticklabels=tokens, 
                yticklabels=tokens,
                cmap='YlOrRd', 
                annot=False,
                cbar_kws={'label': 'Attention Ağırlığı'})
    plt.title('Self-Attention Ağırlıkları', fontsize=14, pad=20)
    plt.xlabel('Key Tokens', fontsize=12)
    plt.ylabel('Query Tokens', fontsize=12)
    plt.tight_layout()
    plt.show()


def compare_generations(prompt, texts_dict, max_display=150):
    """
    Farklı parametrelerle üretilen metinleri karşılaştırır
    
    Args:
        prompt (str): Başlangıç promptu
        texts_dict (dict): {parametre_adı: üretilen_metin} dictionary
        max_display (int): Gösterilecek maksimum karakter sayısı
    """
    print_section("🔍 Metin Üretimi Karşılaştırması", "📊")
    print(f"💬 Prompt: '{prompt}'\n")
    
    for i, (param_name, text) in enumerate(texts_dict.items(), 1):
        # Promptu kaldır
        generated_part = text[len(prompt):].strip()
        display_text = generated_part[:max_display]
        
        if len(generated_part) > max_display:
            display_text += "..."
        
        print(f"{i}. {param_name}")
        print(f"   └─ {display_text}")
        print(f"   📏 Uzunluk: {len(generated_part)} karakter\n")


def calculate_perplexity(model, tokenizer, text, device):
    """
    Bir metin için perplexity (şaşkınlık) değerini hesaplar
    Düşük perplexity = Model metni daha iyi anlıyor
    
    Args:
        model: Dil modeli
        tokenizer: Tokenizer
        text (str): Değerlendirilecek metin
        device: torch.device
    
    Returns:
        float: Perplexity değeri
    """
    encodings = tokenizer(text, return_tensors='pt')
    input_ids = encodings.input_ids.to(device)
    
    with torch.no_grad():
        outputs = model(input_ids, labels=input_ids)
        loss = outputs.loss
        perplexity = torch.exp(loss)
    
    return perplexity.item()


def plot_token_probabilities(probabilities, tokens, top_k=10):
    """
    En yüksek olasılıklı tokenleri görselleştirir
    
    Args:
        probabilities (tensor): Token olasılıkları
        tokens (list): Token listesi
        top_k (int): Gösterilecek top token sayısı
    """
    top_probs, top_indices = torch.topk(probabilities, top_k)
    top_tokens = [tokens[idx] for idx in top_indices]
    
    plt.figure(figsize=(10, 6))
    plt.barh(range(top_k), top_probs.cpu().numpy(), color='steelblue')
    plt.yticks(range(top_k), top_tokens)
    plt.xlabel('Olasılık', fontsize=12)
    plt.title(f'En Olası {top_k} Token', fontsize=14, pad=20)
    plt.gca().invert_yaxis()
    
    # Değerleri çubukların üzerine yaz
    for i, prob in enumerate(top_probs.cpu().numpy()):
        plt.text(prob + 0.001, i, f'{prob:.4f}', va='center')
    
    plt.tight_layout()
    plt.show()


print("✅ Yardımcı fonksiyonlar tanımlandı!")
print("📋 Fonksiyonlar:")
print("   • print_section() - Başlık yazdırma")
print("   • visualize_attention_weights() - Attention görselleştirme")
print("   • compare_generations() - Metin karşılaştırma")
print("   • calculate_perplexity() - Perplexity hesaplama")
print("   • plot_token_probabilities() - Token olasılıkları grafiği")
````

### 2️⃣ 🎯 Attention Mekanizması

<div style="background-color: #fff3e0; padding: 15px; border-left: 5px solid #ff9800; border-radius: 5px;">
    <p><strong>Attention:</strong> Modelin, bir kelimenin diğer kelimelerle olan ilişkisini öğrenmesidir.</p>
</div>

**📌 Örnek Cümle:** *"The cat sat on the mat"*

```
"sat" kelimesi için attention ağırlıkları:

🐱 "cat"  ████████████ 0.85  ← Yüksek (Özne - kim oturdu?)
🪑 "sat"  ██████ 0.45         ← Orta (Kendisi)
📍 "on"   ███ 0.20            ← Orta (Konum bilgisi)
🟦 "mat"  ██████████ 0.75     ← Yüksek (Nesne - nerede?)
📰 "the"  █ 0.05              ← Düşük (Yardımcı kelime)
```

> **🔥 Self-Attention Gücü:** Model, "sat" kelimesini anlamak için "cat" (özne) ve "mat" (nesne) kelimelerine yüksek attention verir!

### 4️⃣ 🎲 Top-k ve Top-p Sampling

<div style="background: linear-gradient(to right, #fa709a 0%, #fee140 100%); padding: 15px; border-radius: 10px; color: white;">
    <h4 style="color: white; margin-top: 0;">Sampling Stratejileri</h4>
    <p>Model, bir sonraki tokeni seçerken kullandığı yöntemlerdir.</p>
</div>

#### 🔸 **Top-k Sampling**

En yüksek **k** olasılıklı tokenden birini seç.

```python
top_k = 50 → En olası 50 token arasından seç

Örnek:
Token Olasılıkları:
1. "amazing" → 0.25  ✅ Seçilebilir (Top 50'de)
2. "great"   → 0.20  ✅ Seçilebilir
...
50. "okay"   → 0.001 ✅ Seçilebilir
51. "weird"  → 0.0009 ❌ Seçilemez (50'den sonra)
```

#### 🔸 **Top-p (Nucleus) Sampling**

Kümülatif olasılığı **p**'ye ulaşan tokenleri seç.

```python
top_p = 0.9 → Kümülatif %90'a ulaşan token kümesinden seç

Örnek:
Token          Olasılık    Kümülatif
1. "amazing"   0.35        0.35  ✅
2. "great"     0.25        0.60  ✅
3. "good"      0.20        0.80  ✅
4. "nice"      0.12        0.92  ❌ (0.9 geçildi)
5. "okay"      0.08        1.00  ❌
```

**📊 Karşılaştırma:**

| Strateji | Avantaj | Dezavantaj | Ne Zaman Kullan? |
|----------|---------|------------|------------------|
| **Greedy** | Deterministik, hızlı | Tekrarlayıcı | Kod üretimi |
| **Top-k** | Kontrollü çeşitlilik | Sabit k her zaman uygun olmaz | Genel kullanım |
| **Top-p** | Dinamik, esnek | Hesaplama maliyeti | Yaratıcı metin |
| **Beam Search** | En iyi global seçim | Çok yavaş | Çeviri, özetleme |

### 🔧 Yardımcı Fonksiyonlar

Bu notebook boyunca kullanacağımız özel fonksiyonları tanımlayalım.

````python
def print_section(title, emoji="📌"):
    """
    Bölüm başlıklarını görsel olarak yazdıran fonksiyon
    
    Args:
        title (str): Başlık metni
        emoji (str): Başlangıç emojisi
    """
    print("\n" + "="*70)
    print(f"{emoji} {title}")
    print("="*70 + "\n")


def visualize_attention_weights(text, attention_scores, tokenizer):
    """
    Attention ağırlıklarını ısı haritası olarak görselleştirir
    
    Args:
        text (str): Girdi metni
        attention_scores (tensor): Attention skorları
        tokenizer: Tokenizer nesnesi
    """
    tokens = tokenizer.tokenize(text)
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(attention_scores.cpu().numpy(), 
                xticklabels=tokens, 
                yticklabels=tokens,
                cmap='YlOrRd', 
                annot=False,
                cbar_kws={'label': 'Attention Ağırlığı'})
    plt.title('Self-Attention Ağırlıkları', fontsize=14, pad=20)
    plt.xlabel('Key Tokens', fontsize=12)
    plt.ylabel('Query Tokens', fontsize=12)
    plt.tight_layout()
    plt.show()


def compare_generations(prompt, texts_dict, max_display=150):
    """
    Farklı parametrelerle üretilen metinleri karşılaştırır
    
    Args:
        prompt (str): Başlangıç promptu
        texts_dict (dict): {parametre_adı: üretilen_metin} dictionary
        max_display (int): Gösterilecek maksimum karakter sayısı
    """
    print_section("🔍 Metin Üretimi Karşılaştırması", "📊")
    print(f"💬 Prompt: '{prompt}'\n")
    
    for i, (param_name, text) in enumerate(texts_dict.items(), 1):
        # Promptu kaldır
        generated_part = text[len(prompt):].strip()
        display_text = generated_part[:max_display]
        
        if len(generated_part) > max_display:
            display_text += "..."
        
        print(f"{i}. {param_name}")
        print(f"   └─ {display_text}")
        print(f"   📏 Uzunluk: {len(generated_part)} karakter\n")


def calculate_perplexity(model, tokenizer, text, device):
    """
    Bir metin için perplexity (şaşkınlık) değerini hesaplar
    Düşük perplexity = Model metni daha iyi anlıyor
    
    Args:
        model: Dil modeli
        tokenizer: Tokenizer
        text (str): Değerlendirilecek metin
        device: torch.device
    
    Returns:
        float: Perplexity değeri
    """
    encodings = tokenizer(text, return_tensors='pt')
    input_ids = encodings.input_ids.to(device)
    
    with torch.no_grad():
        outputs = model(input_ids, labels=input_ids)
        loss = outputs.loss
        perplexity = torch.exp(loss)
    
    return perplexity.item()


def plot_token_probabilities(probabilities, tokens, top_k=10):
    """
    En yüksek olasılıklı tokenleri görselleştirir
    
    Args:
        probabilities (tensor): Token olasılıkları
        tokens (list): Token listesi
        top_k (int): Gösterilecek top token sayısı
    """
    top_probs, top_indices = torch.topk(probabilities, top_k)
    top_tokens = [tokens[idx] for idx in top_indices]
    
    plt.figure(figsize=(10, 6))
    plt.barh(range(top_k), top_probs.cpu().numpy(), color='steelblue')
    plt.yticks(range(top_k), top_tokens)
    plt.xlabel('Olasılık', fontsize=12)
    plt.title(f'En Olası {top_k} Token', fontsize=14, pad=20)
    plt.gca().invert_yaxis()
    
    # Değerleri çubukların üzerine yaz
    for i, prob in enumerate(top_probs.cpu().numpy()):
        plt.text(prob + 0.001, i, f'{prob:.4f}', va='center')
    
    plt.tight_layout()
    plt.show()


print("✅ Yardımcı fonksiyonlar tanımlandı!")
print("📋 Fonksiyonlar:")
print("   • print_section() - Başlık yazdırma")
print("   • visualize_attention_weights() - Attention görselleştirme")
print("   • compare_generations() - Metin karşılaştırma")
print("   • calculate_perplexity() - Perplexity hesaplama")
print("   • plot_token_probabilities() - Token olasılıkları grafiği")
````

## 2.5 🧹 Metin Ön İşleme

<div style="background-color: #fff3e0; padding: 15px; border-left: 5px solid #ff9800; border-radius: 5px;">
    <h3 style="color: #e65100; margin-top: 0;">🧹 Neden Ön İşleme?</h3>
    <p>Metin verisini model için daha uygun hale getirmek için temizleme ve normalizasyon işlemleri yaparız.</p>
</div>

### 🔧 Ön İşleme Adımları:

| Adım | Ne Yapar? | GPT-2 için Gerekli mi? |
|------|-----------|------------------------|
| **1️⃣ Küçük Harfe Çevirme** | "Hello" → "hello" | ❌ (GPT-2 case-sensitive) |
| **2️⃣ HTML Tag Temizleme** | `<br />` → ` ` | ✅ Evet |
| **3️⃣ Özel Karakter Temizleme** | `@#$%` → ` ` | ⚠️ Kısmen |
| **4️⃣ Stopwords Kaldırma** | "the, a, an" → ❌ | ❌ (Dil akıcılığı için gerekli) |
| **5️⃣ Lemmatization** | "running" → "run" | ❌ (GPT-2 tokenizer halleder) |
| **6️⃣ Çok Fazla Boşluk** | `"a    b"` → `"a b"` | ✅ Evet |

> **💡 Önemli:** GPT-2 gibi modern transformer modeller **minimal ön işleme** gerektirir! Tokenizer zaten çok güçlü.

````python
# Metin ön işleme fonksiyonu
import re

def clean_text(text):
    """
    Metin temizleme fonksiyonu (GPT-2 için minimal temizlik)
    
    Args:
        text (str): Ham metin
    
    Returns:
        str: Temizlenmiş metin
    """
    # HTML etiketlerini kaldır (örn: <br />, <p>, vb.)
    text = re.sub(r'<[^>]+>', ' ', text)
    
    # URL'leri kaldır (opsiyonel)
    text = re.sub(r'http\S+|www\S+', '', text)
    
    # Birden fazla boşluğu tek boşluğa indir
    text = re.sub(r'\s+', ' ', text)
    
    # Başındaki ve sonundaki boşlukları temizle
    text = text.strip()
    
    return text


# Örnek uygulama
print_section("🧹 METİN ÖN İŞLEME ÖRNEĞİ", "🔧")

# Ham metin örneği
sample_text = train_data[10]['text']

print("📝 HAM METİN:")
print("="*80)
print(sample_text[:500])
print("\n" + "="*80)

# Temizlenmiş metin
cleaned_text = clean_text(sample_text)

print("\n✨ TEMİZLENMİŞ METİN:")
print("="*80)
print(cleaned_text[:500])
print("\n" + "="*80)

# İstatistikler
print(f"\n📊 İSTATİSTİKLER:")
print(f"   • Ham Metin Uzunluğu: {len(sample_text)} karakter")
print(f"   • Temiz Metin Uzunluğu: {len(cleaned_text)} karakter")
print(f"   • Fark: {len(sample_text) - len(cleaned_text)} karakter temizlendi")

print("\n✅ Metin temizleme fonksiyonu hazır!")
````

````python
# Veri setini temizle ve DataFrame'e dönüştür (opsiyonel ama kullanışlı)
print_section("💾 VERİ SETİNİ DATAFRAME'E DÖNÜŞTÜRME", "🐼")

# Eğitim verisinden DataFrame oluştur (ilk 1000 örnek - hız için)
sample_size = 1000
train_df = pd.DataFrame({
    'text': [train_data[i]['text'] for i in range(sample_size)],
    'label': [train_data[i]['label'] for i in range(sample_size)]
})

# Temiz metin sütunu ekle
train_df['cleaned_text'] = train_df['text'].apply(clean_text)

# Metin uzunlukları ekle
train_df['char_length'] = train_df['cleaned_text'].apply(len)
train_df['word_count'] = train_df['cleaned_text'].apply(lambda x: len(x.split()))

# Sentiment etiketini anlamlı hale getir
train_df['sentiment'] = train_df['label'].map({0: 'Negative', 1: 'Positive'})

print("✅ DataFrame oluşturuldu!")
print(f"\n📊 DataFrame Boyutu: {train_df.shape}")
print(f"   • Satır: {train_df.shape[0]:,}")
print(f"   • Sütun: {train_df.shape[1]}")

print("\n📋 İLK 5 SATIR:")
print("="*80)
display(train_df[['sentiment', 'char_length', 'word_count', 'cleaned_text']].head())

print("\n📈 ÖZET İSTATİSTİKLER:")
print("="*80)
display(train_df[['char_length', 'word_count']].describe())

print("\n✅ Veri hazırlığı tamamlandı! Artık GPT-2 ile çalışmaya hazırız.")
````

## 3.2 🔤 Tokenization - Derinlemesine İnceleme

    <h3 style="color: white; margin-top: 0;">🔑 Tokenization Nedir?</h3>
    <p style="font-size: 16px; line-height: 1.8;">
        <strong>Tokenization</strong>, metni modelin anlayabileceği sayısal tokenlere (ID'lere) dönüştürme işlemidir.
        GPT-2, <strong>Byte-Pair Encoding (BPE)</strong> algoritmasını kullanır.

### 📝 BPE Nasıl Çalışır?

<div style="background-color: #f5f5f5; padding: 20px; border-radius: 10px; margin: 20px 0;">
    <strong>Örnek Metin:</strong> <code>"ChatGPT is amazing!"</code>
    <br><br>
    <strong>🔽 Tokenization Süreci:</strong>
    <pre style="background: #fff; padding: 15px; border-radius: 8px; border-left: 4px solid #2196F3; overflow-x: auto;">
Ham Metin    → "ChatGPT is amazing!"
                    ↓
Karakter     → ['C','h','a','t','G','P','T',' ','i','s',' ','a','m','a','z','i','n','g','!']
                    ↓
BPE Tokenler → ["Chat", "G", "PT", " is", " amazing", "!"]
                    ↓
Token IDs    → [30813, 38, 11571, 318, 4998, 0]
    </pre>
</div>

<div style="background-color: #e8f5e9; padding: 15px; border-left: 5px solid #4caf50; border-radius: 8px; margin: 20px 0;">
    <strong>💡 Önemli Noktalar:</strong>
    <ul style="margin: 10px 0;">
        <li>🔸 <strong>Sık kullanılan kelimeler</strong> → 1 token (örn: "the", "is")</li>
        <li>🔸 <strong>Nadir kelimeler</strong> → Birden fazla token (örn: "ChatGPT" → ["Chat", "G", "PT"])</li>
        <li>🔸 <strong>Boşluklar önemli!</strong> "is" ≠ " is" (başında boşluk olan farklı token)</li>
        <li>🔸 <strong>Büyük/küçük harf</strong> → Farklı tokenler ("The" ≠ "the")</li>
    </ul>
</div>

---

````python
# Tokenization örnekleri ve detaylı analiz
print_section("🔤 TOKENIZATION ÖRNEKLERİ", "🔬")

# Farklı metinler deneyelim
test_texts = [
    "Hello, how are you?",
    "The movie was amazing!",
    "ChatGPT is a powerful AI model.",
    "This film is absolutely fantastic and I loved every moment of it!",
    "Bad movie. Waste of time."
]

print("📝 ÇEŞİTLİ METİNLERİN TOKENIZATION ANALİZİ:\n")

for i, text in enumerate(test_texts, 1):
    print(f"{i}. 📄 Metin: \"{text}\"")
    print("   " + "="*70)
    
    # Tokenize et
    tokens = tokenizer.tokenize(text)
    token_ids = tokenizer.encode(text)
    
    # Bilgileri göster
    print(f"   🔤 Tokenler: {tokens}")
    print(f"   🔢 Token IDs: {token_ids}")
    print(f"   📊 Token Sayısı: {len(tokens)}")
    print(f"   📏 Karakter Sayısı: {len(text)}")
    print(f"   📐 Token/Karakter Oranı: {len(tokens)/len(text):.2f}")
    
    # Decode işlemi - token'ları tekrar metne çevir
    decoded_text = tokenizer.decode(token_ids)
    print(f"   ✅ Decoded Metin: \"{decoded_text}\"")
    print(f"   🔍 Orijinal = Decoded? {text == decoded_text}")
    print()

# Özel durum örnekleri
print("\n" + "="*70)
print("🎯 ÖZEL DURUMLAR:\n")

# 1. Boşluk farkı
text1 = "is"
text2 = " is"
print(f"1️⃣ Boşluk Farkı:")
print(f"   '{text1}' → Token IDs: {tokenizer.encode(text1)}")
print(f"   '{text2}' → Token IDs: {tokenizer.encode(text2)}")
print(f"   💡 Başta boşluk olan farklı token!\n")

# 2. Büyük/küçük harf farkı
text3 = "The"
text4 = "the"
print(f"2️⃣ Büyük/Küçük Harf Farkı:")
print(f"   '{text3}' → Token IDs: {tokenizer.encode(text3)}")
print(f"   '{text4}' → Token IDs: {tokenizer.encode(text4)}")
print(f"   💡 Büyük harf farklı token!\n")

# 3. Nadir kelime
text5 = "supercalifragilisticexpialidocious"
tokens5 = tokenizer.tokenize(text5)
print(f"3️⃣ Nadir Kelime:")
print(f"   '{text5}'")
print(f"   → Tokenler: {tokens5}")
print(f"   → {len(tokens5)} tokene ayrıldı!")
print(f"   💡 Nadir kelimeler birden fazla tokene bölünür!\n")

print("✅ Tokenization detayları anlaşıldı!")
````

## 3.5 🎲 Top-k ve Top-p Sampling

**Sampling Stratejileri:**
- **Top-k:** En yüksek olasılıklı k token arasından seçim yapar. Kötü seçenekler elenir.
- **Top-p (Nucleus):** Kümülatif olasılığı p'ye ulaşan tokenleri seçer. Dinamik ve daha çeşitli sonuçlar üretir.

**Karşılaştırma:**
| Özellik         | Top-k           | Top-p           |
|-----------------|-----------------|-----------------|
| Seçim Mantığı   | Sabit k token   | Dinamik token   |
| Esneklik        | Sabit           | Uyarlanabilir   |
| Hesaplama      | Hızlı           | Biraz yavaş     |
| Kullanım        | Genel amaçlı    | Yaratıcı işler  |

**Pro İpucu:** Genellikle Top-p (0.9) + Temperature (0.7-1.0) en iyi sonucu verir!

## 3.6 🎨 Prompt Engineering - Etkili Prompt Yazma

**Prompt Engineering Nedir?**
- Modelden istediğiniz çıktıyı almak için başlangıç metnini (prompt) stratejik olarak tasarlama sanatıdır.

**Başlıca Prompt Türleri:**
- Basit: "The movie was"
- Context: "Movie Review: This film was"
- Instructional: "Write a positive movie review:"
- Role-based: "As a professional film critic, I must say that this movie"
- Format: "Movie Review Summary:\n- Acting:\n- Plot:\n- Overall: This movie was"
- Constraint: "In exactly 2 sentences, describe why this movie is amazing: This film"

**İyi Prompt İlkeleri:**
- Açık ve spesifik olun
- Bağlam sağlayın
- Format belirtin
- Ton ve stil ekleyin

**İleri Teknikler:**
- Chain-of-Thought: Adım adım düşünmeye teşvik edin
- Role Prompting: Modele rol verin
- Constraint Prompting: Kısıtlamalar ekleyin

````python
# Top-k ve Top-p parametreleri ile deney
print_section("🎲 TOP-K VE TOP-P DENEYLERİ", "🧪")

prompt = "The best thing about artificial intelligence is that"

print(f"💬 Prompt: \"{prompt}\"\n")
print("🧪 Farklı sampling stratejileri deneniyor...\n")
print("="*80)

# Farklı konfigürasyonlar
configs = [
    {"name": "Greedy (No Sampling)", "do_sample": False, "top_k": None, "top_p": None, "temp": 1.0},
    {"name": "Top-k=10", "do_sample": True, "top_k": 10, "top_p": 1.0, "temp": 0.8},
    {"name": "Top-k=50 (Default)", "do_sample": True, "top_k": 50, "top_p": 1.0, "temp": 0.8},
    {"name": "Top-p=0.7 (Nucleus)", "do_sample": True, "top_k": 0, "top_p": 0.7, "temp": 0.8},
    {"name": "Top-p=0.95", "do_sample": True, "top_k": 0, "top_p": 0.95, "temp": 0.8},
    {"name": "Kombine (k=50, p=0.9)", "do_sample": True, "top_k": 50, "top_p": 0.9, "temp": 0.8},
]

results_sampling = {}

for config in configs:
    print(f"\n🔧 {config['name']}")
    print("-"*80)
    
    # Tokenize
    input_ids = tokenizer.encode(prompt, return_tensors='pt').to(device)
    
    # Generate parametreleri
    gen_params = {
        'input_ids': input_ids,
        'max_length': 70,
        'num_return_sequences': 1,
        'pad_token_id': tokenizer.eos_token_id,
        'do_sample': config['do_sample']
    }
    
    # Opsiyonel parametreler ekle
    if config['temp'] is not None:
        gen_params['temperature'] = config['temp']
    if config['top_k'] is not None and config['top_k'] > 0:
        gen_params['top_k'] = config['top_k']
    if config['top_p'] is not None and config['top_p'] < 1.0:
        gen_params['top_p'] = config['top_p']
    
    # Üret
    with torch.no_grad():
        output = model.generate(**gen_params)
    
    # Decode
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    results_sampling[config['name']] = generated_text
    
    # Yeni kısmı göster
    new_part = generated_text[len(prompt):].strip()
    print(f"📝 {new_part[:180]}")
    if len(new_part) > 180:
        print(f"   ... (toplam {len(new_part)} karakter)")
    print()

print("="*80)

# Karşılaştırmalı özet
print("\n📊 KARŞILAŞTIRMALI ÖZET:\n")

for i, (name, text) in enumerate(results_sampling.items(), 1):
    new_part = text[len(prompt):].strip()
    print(f"{i}. {name}")
    print(f"   └─ {new_part[:120]}{'...' if len(new_part) > 120 else ''}")
    print(f"   📏 {len(new_part)} karakter\n")

# Çeşitlilik analizi
print("="*80)
print("\n💡 GÖZLEMLER:")
print("   • Greedy: Her zaman aynı sonuç (deterministik)")
print("   • Top-k=10: Sınırlı çeşitlilik, güvenli seçimler")
print("   • Top-k=50: Daha fazla çeşitlilik")
print("   • Top-p: Dinamik, context'e uygun")
print("   • Kombine: En dengeli yaklaşım (önerilen)")
print("\n✅ Sampling deneyleri tamamlandı!")
````

### 🔧 Yardımcı Fonksiyonlar

Bu notebook boyunca kullanacağımız özel fonksiyonları tanımlayalım.

````python
def print_section(title, emoji="📌"):
    """
    Bölüm başlıklarını görsel olarak yazdıran fonksiyon
    
    Args:
        title (str): Başlık metni
        emoji (str): Başlangıç emojisi
    """
    print("\n" + "="*70)
    print(f"{emoji} {title}")
    print("="*70 + "\n")


def visualize_attention_weights(text, attention_scores, tokenizer):
    """
    Attention ağırlıklarını ısı haritası olarak görselleştirir
    
    Args:
        text (str): Girdi metni
        attention_scores (tensor): Attention skorları
        tokenizer: Tokenizer nesnesi
    """
    tokens = tokenizer.tokenize(text)
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(attention_scores.cpu().numpy(), 
                xticklabels=tokens, 
                yticklabels=tokens,
                cmap='YlOrRd', 
                annot=False,
                cbar_kws={'label': 'Attention Ağırlığı'})
    plt.title('Self-Attention Ağırlıkları', fontsize=14, pad=20)
    plt.xlabel('Key Tokens', fontsize=12)
    plt.ylabel('Query Tokens', fontsize=12)
    plt.tight_layout()
    plt.show()


def compare_generations(prompt, texts_dict, max_display=150):
    """
    Farklı parametrelerle üretilen metinleri karşılaştırır
    
    Args:
        prompt (str): Başlangıç promptu
        texts_dict (dict): {parametre_adı: üretilen_metin} dictionary
        max_display (int): Gösterilecek maksimum karakter sayısı
    """
    print_section("🔍 Metin Üretimi Karşılaştırması", "📊")
    print(f"💬 Prompt: '{prompt}'\n")
    
    for i, (param_name, text) in enumerate(texts_dict.items(), 1):
        # Promptu kaldır
        generated_part = text[len(prompt):].strip()
        display_text = generated_part[:max_display]
        
        if len(generated_part) > max_display:
            display_text += "..."
        
        print(f"{i}. {param_name}")
        print(f"   └─ {display_text}")
        print(f"   📏 Uzunluk: {len(generated_part)} karakter\n")


def calculate_perplexity(model, tokenizer, text, device):
    """
    Bir metin için perplexity (şaşkınlık) değerini hesaplar
    Düşük perplexity = Model metni daha iyi anlıyor
    
    Args:
        model: Dil modeli
        tokenizer: Tokenizer
        text (str): Değerlendirilecek metin
        device: torch.device
    
    Returns:
        float: Perplexity değeri
    """
    encodings = tokenizer(text, return_tensors='pt')
    input_ids = encodings.input_ids.to(device)
    
    with torch.no_grad():
        outputs = model(input_ids, labels=input_ids)
        loss = outputs.loss
        perplexity = torch.exp(loss)
    
    return perplexity.item()


def plot_token_probabilities(probabilities, tokens, top_k=10):
    """
    En yüksek olasılıklı tokenleri görselleştirir
    
    Args:
        probabilities (tensor): Token olasılıkları
        tokens (list): Token listesi
        top_k (int): Gösterilecek top token sayısı
    """
    top_probs, top_indices = torch.topk(probabilities, top_k)
    top_tokens = [tokens[idx] for idx in top_indices]
    
    plt.figure(figsize=(10, 6))
    plt.barh(range(top_k), top_probs.cpu().numpy(), color='steelblue')
    plt.yticks(range(top_k), top_tokens)
    plt.xlabel('Olasılık', fontsize=12)
    plt.title(f'En Olası {top_k} Token', fontsize=14, pad=20)
    plt.gca().invert_yaxis()
    
    # Değerleri çubukların üzerine yaz
    for i, prob in enumerate(top_probs.cpu().numpy()):
        plt.text(prob + 0.001, i, f'{prob:.4f}', va='center')
    
    plt.tight_layout()
    plt.show()


print("✅ Yardımcı fonksiyonlar tanımlandı!")
print("📋 Fonksiyonlar:")
print("   • print_section() - Başlık yazdırma")
print("   • visualize_attention_weights() - Attention görselleştirme")
print("   • compare_generations() - Metin karşılaştırma")
print("   • calculate_perplexity() - Perplexity hesaplama")
print("   • plot_token_probabilities() - Token olasılıkları grafiği")
````

<div style="background-color: #fff3e0; padding: 15px; border-left: 5px solid #ff9800; border-radius: 5px;">
    <p><strong>Attention:</strong> Modelin, bir kelimenin diğer kelimelerle olan ilişkisini öğrenmesidir.</p>
</div>

**📌 Örnek Cümle:** *"The cat sat on the mat"*

```
"sat" kelimesi için attention ağırlıkları:

🐱 "cat"  ████████████ 0.85  ← Yüksek (Özne - kim oturdu?)
🪑 "sat"  ██████ 0.45         ← Orta (Kendisi)
📍 "on"   ███ 0.20            ← Orta (Konum bilgisi)
🟦 "mat"  ██████████ 0.75     ← Yüksek (Nesne - nerede?)
📰 "the"  █ 0.05              ← Düşük (Yardımcı kelime)
```

> **🔥 Self-Attention Gücü:** Model, "sat" kelimesini anlamak için "cat" (özne) ve "mat" (nesne) kelimelerine yüksek attention verir!

### 4️⃣ 🎲 Top-k ve Top-p Sampling

<div style="background: linear-gradient(to right, #fa709a 0%, #fee140 100%); padding: 15px; border-radius: 10px; color: white;">
    <h4 style="color: white; margin-top: 0;">Sampling Stratejileri</h4>
    <p>Model, bir sonraki tokeni seçerken kullandığı yöntemlerdir.</p>
</div>

#### 🔸 **Top-k Sampling**

En yüksek **k** olasılıklı tokenden birini seç.

```python
top_k = 50 → En olası 50 token arasından seç

Örnek:
Token Olasılıkları:
1. "amazing" → 0.25  ✅ Seçilebilir (Top 50'de)
2. "great"   → 0.20  ✅ Seçilebilir
...
50. "okay"   → 0.001 ✅ Seçilebilir
51. "weird"  → 0.0009 ❌ Seçilemez (50'den sonra)
```

#### 🔸 **Top-p (Nucleus) Sampling**

Kümülatif olasılığı **p**'ye ulaşan tokenleri seç.

```python
top_p = 0.9 → Kümülatif %90'a ulaşan token kümesinden seç

Örnek:
Token          Olasılık    Kümülatif
1. "amazing"   0.35        0.35  ✅
2. "great"     0.25        0.60  ✅
3. "good"      0.20        0.80  ✅
4. "nice"      0.12        0.92  ❌ (0.9 geçildi)
5. "okay"      0.08        1.00  ❌
```

**📊 Karşılaştırma:**

| Strateji | Avantaj | Dezavantaj | Ne Zaman Kullan? |
|----------|---------|------------|------------------|
| **Greedy** | Deterministik, hızlı | Tekrarlayıcı | Kod üretimi |
| **Top-k** | Kontrollü çeşitlilik | Sabit k her zaman uygun olmaz | Genel kullanım |
| **Top-p** | Dinamik, esnek | Hesaplama maliyeti | Yaratıcı metin |
| **Beam Search** | En iyi global seçim | Çok yavaş | Çeviri, özetleme |

- Context: "Movie Review: This film was"
- Instructional: "Write a positive movie review:"
- Role-based: "As a professional film critic, I must say that this movie"
- Format: "Movie Review Summary:\n- Acting:\n- Plot:\n- Overall: This movie was"
- Constraint: "In exactly 2 sentences, describe why this movie is amazing: This film"

**İyi Prompt İlkeleri:**
- Açık ve spesifik olun
- Bağlam sağlayın
- Format belirtin
- Ton ve stil ekleyin

**İleri Teknikler:**
- Chain-of-Thought: Adım adım düşünmeye teşvik edin
- Role Prompting: Modele rol verin
- Constraint Prompting: Kısıtlamalar ekleyin

````python
# Farklı prompt stratejileri ile deney
print_section("🎨 PROMPT ENGINEERING DENEYLERİ", "✨")

# IMDb veri setinden rastgele pozitif bir yorum alalım (ilham için)
sample_review = train_data[42]['text'][:100]  # İlk 100 karakter
print(f"📚 Gerçek IMDb Yorumu (örnek): \"{sample_review}...\"\n")
print("="*80)

# Farklı prompt stratejileri
prompts = {
    "1. Basit": "This movie was",
    
    "2. Context": "IMDb Review (Rating: 9/10):\nThis movie was",
    
    "3. Instructional": "Write a positive and detailed movie review:\n\nThis film",
    
    "4. Role-Based": "As a professional film critic, I must say that this movie",
    
    "5. Format": "Movie Review Summary:\n- Acting:\n- Plot:\n- Overall: This movie was",
    
    "6. Constraint": "In exactly 2 sentences, describe why this movie is amazing: This film"
}

print("\n🎨 Farklı prompt stratejileri deneniyor...\n")

results_prompts = {}

for name, prompt in prompts.items():
    print(f"\n{name}")
    print("-"*80)
    print(f"💬 Prompt: \"{prompt[:80]}{'...' if len(prompt) > 80 else ''}\"")
    
    # Tokenize
    input_ids = tokenizer.encode(prompt, return_tensors='pt').to(device)
    
    # Generate
    with torch.no_grad():
        output = model.generate(
            input_ids,
            max_length=input_ids.shape[1] + 60,  # Prompt + 60 token
            temperature=0.8,
            do_sample=True,
            top_k=50,
            top_p=0.95,
            num_return_sequences=1,
            pad_token_id=tokenizer.eos_token_id
        )
    
    # Decode
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    results_prompts[name] = generated_text
    
    # Yeni kısmı göster
    new_part = generated_text[len(prompt):].strip()
    print(f"📝 Üretilen: {new_part[:150]}")
    if len(new_part) > 150:
        print(f"            ... (toplam {len(new_part)} karakter)")
    print()

print("="*80)

# Karşılaştırmalı analiz
print("\n📊 KARŞILAŞTIRMALI ANALİZ:\n")

for name, text in results_prompts.items():
    # Prompt'u bul ve kaldır
    for prompt_name, prompt_text in prompts.items():
        if prompt_name == name:
            new_part = text[len(prompt_text):].strip()
            break
    
    print(f"{name}")
    print(f"  Prompt tipi: {name.split('.')[1].strip()}")
    print(f"  Üretilen: {new_part[:100]}{'...' if len(new_part) > 100 else ''}")
    print(f"  Uzunluk: {len(new_part)} karakter")
    print()

# Önemli gözlemler
print("="*80)
print("\n💡 ÖNEMLI GÖZLEMLER:")
print("   ✅ Context verilen promptlar daha tutarlı sonuçlar üretiyor")
print("   ✅ Role-based promptlar daha profesyonel ton kullanıyor")
print("   ✅ Format belirtilen promptlar yapılandırılmış çıktı veriyor")
print("   ✅ Instructional promptlar daha uzun ve detaylı içerik üretiyor")
print("\n🎯 Sonuç: Prompt tasarımı sonucu dramatik şekilde etkiliyor!")
print("\n✅ Prompt engineering deneyleri tamamlandı!")
````
