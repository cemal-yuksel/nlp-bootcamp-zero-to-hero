## ❓ Word Sense Disambiguation (WSD) Nedir?

> **🎯 Tanım**
> 
> **Word Sense Disambiguation (Kelime Anlamı Belirsizliği Giderme)**, bir kelimenin belirli bir bağlamda hangi anlamda kullanıldığını otomatik olarak belirleme işlemidir. Doğal dil işlemede **temel** ve **kritik** bir görevdir.

### 🧩 Basit Bir Örnek

**"Bank"** kelimesi İngilizce'de birden fazla anlama sahiptir:

| 🏦 **Anlam 1: Finans Kurumu** | 🌊 **Anlam 2: Nehir Kıyısı** |
|-------------------------------|-------------------------------|
| "I went to the **bank** to deposit money." | "The river **bank** was flooded after rain." |
| 💰 Para yatırmak için bankaya gittim. | 🏞️ Yağmurdan sonra nehir kıyısı su bastı. |

### 🎯 WSD'nin Görevi

WSD algoritmaları, **cümlenin bağlamını** (context) analiz ederek **"bank"** kelimesinin hangi anlamda kullanıldığını otomatik olarak belirler.

---

## 🎪 Daha Fazla Gerçek Hayat Örneği

### 📚 İngilizce Örnekler

| 🔤 Kelime | 📝 Cümle | 💡 Anlam |
|-----------|----------|---------|
| 🌿 **PLANT** | We need to **plant** more trees. | 🌱 Dikmek (fiil) |
| 🌿 **PLANT** | The **plant** needs water. | 🪴 Bitki (isim) |

| 🏃 **RUN** | The program will **run** automatically. | 💻 Çalıştırmak (fiil) |

## 🧠 WSD Neden Önemlidir?

> ⚠️ **Önemli!**
>
> WSD, birçok NLP uygulamasının **temelini** oluşturur. Doğru anlam belirleme olmadan, makine çevirisi, bilgi çıkarımı, soru-cevap sistemleri gibi uygulamalar düzgün çalışamaz.

### 🎯 WSD'nin Kullanım Alanları

#### 🌍 1. Makine Çevirisi
Kelimelerin farklı dillerdeki doğru karşılıklarını bulmak için WSD kritiktir.
- **Örnek:** "Bank" → TR: "Banka" veya "Kıyı"?

#### 🔍 2. Bilgi Çıkarımı
Metinlerden doğru bilgiyi çıkarmak için kelimelerin anlamlarını bilmek şarttır.
- **Örnek:** "Apple released a new product" → Şirket mi, meyve mi?

#### ❓ 3. Soru-Cevap Sistemleri
Soruları doğru anlamak ve ilgili cevapları bulmak için WSD gereklidir.
- **Örnek:** "What is Java?" → Programlama dili mi, ada mı?

#### 📊 4. Metin Analizi
Sentiment analizi ve konu modelleme gibi görevlerde kelimelerin anlamları önemlidir.
- **Örnek:** "This mouse is broken" → Bilgisayar faresi mi, hayvan mı?

---

## 📊 WSD Problem Türleri

### 🎯 1. Lexical Sample WSD
- 📌 **Tanım:** Belirli bir kelime listesi için anlam belirleme
- 🎯 **Örnek:** Sadece "bank", "plant", "match" kelimelerini çözmek
- ✅ **Avantaj:** Odaklanmış ve daha kolay
- ❌ **Dezavantaj:** Sınırlı kapsam

### 🎯 2. All-Words WSD
- 📌 **Tanım:** Metindeki TÜM kelimelerin anlamlarını belirleme
- 🎯 **Örnek:** Bir paragraftaki her kelimenin anlamını bulmak
- ✅ **Avantaj:** Kapsamlı analiz
- ❌ **Dezavantaj:** Çok daha zor ve maliyetli

---

## 💡 FLASHCARDS - Temel Kavramlar

### 📚 SYNSET
**Synonym Set** - Aynı kavramı temsil eden kelimelerin kümesi. WordNet'in temel yapı taşıdır.
- **Örnek:** {car, auto, automobile, motorcar}

### 🎯 CONTEXT
Bir kelimenin anlamını belirlemek için kullanılan çevresindeki kelimeler ve cümle yapısı.
- **Örnek:** "deposit money" → "bank" = finans kurumu

### 🔍 LESK ALGORITHM
Context'teki kelimeleri synset tanımlarıyla karşılaştırarak doğru anlamı bulan klasik WSD algoritması.
- **Temel:** Overlap (örtüşme) hesaplama

### 📖 WORDNET
İngilizce için geliştirilmiş büyük bir lexical veri tabanı. Synset'leri ve aralarındaki ilişkileri içerir.
- **İçerik:** 117,000+ synset, 150,000+ kelime

---

## 🎭 WSD Zorlukları

> ⚠️ **Ana Zorluklar**

- **🔄 Polysemy (Çok Anlamlılık):** Bir kelimenin birden fazla ilişkili anlamı olması
- **🎭 Homonymy (Eş Seslilik):** Aynı yazılıp farklı anlamları olan kelimeler
- **📊 Veri Yetersizliği:** Bazı anlamlar için yeterli eğitim verisi olmaması
- **🌍 Dil ve Domain Farklılıkları:** Her dilde ve alanda farklı kullanımlar
- **🎯 Granularity:** Anlamların ne kadar detaylı ayrılacağı sorunu
- **🔀 Context Window:** Ne kadar bağlam kullanılacağı belirsizliği

## 📖 WordNet Nedir?

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 25px; border-radius: 12px; color: white; margin: 20px 0; box-shadow: 0 8px 25px rgba(0,0,0,0.2);">

        İngilizce kelimeleri <b>synset'ler</b> (synonym sets - eş anlamlı kelime kümeleri) halinde organize eder.

### 🔑 Anahtar Kavramlar

<div style="display: grid; grid-template-columns: repeat(2, 1fr); gap: 20px; margin: 25px 0;">
    <div style="background-color: #dbeafe; padding: 20px; border-radius: 10px; border-left: 5px solid #3b82f6;">
        <h4 style="color: #1e40af; margin-top: 0;">🎯 Synset</h4>
        <p style="color: #1e3a8a; line-height: 1.6;">
            Aynı anlama gelen kelimelerin kümesi. Her synset bir kavramı temsil eder.
        </p>
    </div>
    <div style="background-color: #fce7f3; padding: 20px; border-radius: 10px; border-left: 5px solid #ec4899;">
        <h4 style="color: #9f1239; margin-top: 0;">📝 Lemma</h4>
        <p style="color: #be123c; line-height: 1.6;">
            Bir synset içindeki bireysel kelime. Aynı kelime farklı synset'lerde bulunabilir.
        </p>
    </div>
    <div style="background-color: #dcfce7; padding: 20px; border-radius: 10px; border-left: 5px solid #22c55e;">
        <h4 style="color: #15803d; margin-top: 0;">💬 Definition</h4>
        <p style="color: #166534; line-height: 1.6;">
            Synset'in anlamını açıklayan tanım (gloss). WSD'de context ile karşılaştırılır.
        </p>
    </div>
    <div style="background-color: #fef3c7; padding: 20px; border-radius: 10px; border-left: 5px solid #f59e0b;">
        <h4 style="color: #92400e; margin-top: 0;">📚 Examples</h4>
        <p style="color: #78350f; line-height: 1.6;">
            Synset'in kullanım örnekleri. Kelimenin o anlamdaki kullanımını gösterir.
        </p>
    </div>
</div>

```python
# 🔍 WordNet'i Keşfedelim - "Bank" Kelimesi Örneği

print("🔍 WordNet'te 'BANK' kelimesinin anlamlarını keşfediyoruz...\n")
print("=" * 80)

# "bank" kelimesinin tüm synset'lerini alalım
word = "bank"
synsets = wn.synsets(word)

print(f"📊 '{word.upper()}' kelimesi için toplam {len(synsets)} farklı synset bulundu!\n")

# Her synset'i detaylı inceleyelim
for i, synset in enumerate(synsets, 1):
    print(f"\n{'='*80}")
    print(f"🎯 SYNSET {i}: {synset.name()}")
    print(f"{'='*80}")
    
    # Synset bilgileri
    print(f"📝 POS (Part of Speech): {synset.pos()}")
    
    # POS açıklaması
    pos_names = {
        'n': '🏷️  Noun (İsim)',
        'v': '⚡ Verb (Fiil)',
        'a': '✨ Adjective (Sıfat)',
        'r': '🔄 Adverb (Zarf)'
    }
    print(f"   └─ {pos_names.get(synset.pos(), 'Bilinmiyor')}")
    
    # Tanım (Definition)
    print(f"\n💬 Tanım:")
    print(f"   {synset.definition()}")
    
    # Eş anlamlı kelimeler (Lemmas)
    lemma_names = synset.lemma_names()
    print(f"\n🔤 Eş Anlamlı Kelimeler (Lemmas): {len(lemma_names)} adet")
    print(f"   {', '.join(lemma_names)}")
    
    # Örnek cümleler
    examples = synset.examples()
    if examples:
        print(f"\n📚 Örnek Cümleler:")
        for j, example in enumerate(examples, 1):
            print(f"   {j}. \"{example}\"")
    else:
        print(f"\n📚 Örnek Cümleler: Yok")
    
    # Lexname (kelime kategorisi)
    print(f"\n🏷️  Kategori (Lexname): {synset.lexname()}")

print("\n" + "=" * 80)
print("\n✅ WordNet keşfi tamamlandı!")
print("\n💡 İpucu: Her synset farklı bir anlam/kavram temsil eder.")
```

```python
# 📊 WordNet'te Çok Anlamlı Kelimeleri Görselleştirelim

# Örnek kelimeler ve synset sayıları
ambiguous_words = ['bank', 'plant', 'match', 'run', 'play', 'light', 'bat', 'bark', 'club', 'date']

# Her kelime için synset sayısını hesapla
word_synset_counts = {}
for word in ambiguous_words:
    synsets = wn.synsets(word)
    word_synset_counts[word] = len(synsets)

# Sıralı DataFrame oluştur
df_words = pd.DataFrame({
    'Kelime': list(word_synset_counts.keys()),
    'Synset Sayısı': list(word_synset_counts.values())
}).sort_values('Synset Sayısı', ascending=False)

# Görselleştirme
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Bar chart
colors = plt.cm.viridis(np.linspace(0, 1, len(df_words)))
bars = axes[0].barh(df_words['Kelime'], df_words['Synset Sayısı'], color=colors, edgecolor='black', linewidth=1.5)
axes[0].set_xlabel('Synset Sayısı', fontsize=13, fontweight='bold')
axes[0].set_title('🎯 Çok Anlamlı Kelimelerin Synset Sayıları', fontsize=15, fontweight='bold', pad=20)
axes[0].grid(axis='x', alpha=0.3, linestyle='--')

# Bar üzerine değerleri ekle
for i, (bar, value) in enumerate(zip(bars, df_words['Synset Sayısı'])):
    axes[0].text(value + 0.3, bar.get_y() + bar.get_height()/2, str(value), 
                va='center', fontweight='bold', fontsize=11)

# Pie chart
colors_pie = plt.cm.Set3(np.linspace(0, 1, len(df_words)))
wedges, texts, autotexts = axes[1].pie(df_words['Synset Sayısı'], 
                                        labels=df_words['Kelime'],
                                        autopct='%1.1f%%',
                                        startangle=90,
                                        colors=colors_pie,
                                        explode=[0.05] * len(df_words),
                                        shadow=True)

# Pie chart text formatı
for text in texts:
    text.set_fontsize(11)
    text.set_fontweight('bold')
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontsize(10)
    autotext.set_fontweight('bold')

axes[1].set_title('📊 Synset Dağılımı', fontsize=15, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()

# İstatistikler
print("\n📈 İSTATİSTİKLER")
print("=" * 60)
print(f"📊 Toplam kelime sayısı: {len(df_words)}")
print(f"🔢 Toplam synset sayısı: {df_words['Synset Sayısı'].sum()}")
print(f"📈 Ortalama synset sayısı: {df_words['Synset Sayısı'].mean():.2f}")
print(f"🏆 En çok anlamlı kelime: '{df_words.iloc[0]['Kelime']}' ({df_words.iloc[0]['Synset Sayısı']} synset)")
print(f"🥈 İkinci sıra: '{df_words.iloc[1]['Kelime']}' ({df_words.iloc[1]['Synset Sayısı']} synset)")
print("=" * 60)
```

## 🔬 Lesk Algoritması Nedir?

<div style="background: linear-gradient(135deg, #4facfe 0%, #00f2fe 100%); padding: 25px; border-radius: 12px; color: white; margin: 20px 0; box-shadow: 0 8px 25px rgba(0,0,0,0.2);">
    <h3 style="margin: 0 0 15px 0; font-size: 24px;">🎯 Lesk Algoritması</h3>
    <p style="font-size: 16px; line-height: 1.8; margin: 0;">
        <b>Lesk Algoritması</b>, 1986'da Michael Lesk tarafından geliştirilmiş klasik bir WSD yöntemidir.
        Temel mantığı: Bir kelimenin doğru anlamı, <b>context'teki kelimelerle</b> synset tanımları arasında
        <b>en fazla örtüşme (overlap)</b> olan anlamdır.
    </p>
</div>

### 🧮 Algoritma Mantığı

<div style="background-color: #f0f9ff; padding: 20px; border-radius: 10px; margin: 20px 0; border-left: 6px solid #3b82f6;">
    <h4 style="color: #1e40af; margin-top: 0;">📐 Matematiksel Formül</h4>
    <p style="color: #1e3a8a; font-size: 15px; line-height: 1.8;">
        <b>Overlap Score = |Context ∩ Definition|</b>
    </p>
    <p style="color: #1e3a8a; font-size: 14px; line-height: 1.7; margin-top: 15px;">
        📌 <b>Context:</b> Hedef kelimenin etrafındaki kelimeler<br>
        📌 <b>Definition:</b> Synset'in tanımı<br>
        📌 <b>∩ (Kesişim):</b> Her iki kümede de bulunan ortak kelimeler<br>
        🎯 <b>Sonuç:</b> En yüksek overlap skoruna sahip synset seçilir
    </p>
</div>

```python
# 🔬 Simple Lesk Algoritması - Manuel İmplementasyon

def manual_simple_lesk(context_sentence, target_word):
    """
    Simple Lesk algoritmasını manuel olarak uygular
    
    Args:
        context_sentence (str): Hedef kelimenin bulunduğu cümle
        target_word (str): Anlamını bulmak istediğimiz kelime
    
    Returns:
        tuple: (best_synset, best_score, all_scores)
    """
    # 1. Cümleyi tokenize et ve küçük harfe çevir
    context = set(word_tokenize(context_sentence.lower()))
    
    # 2. Hedef kelimenin tüm synset'lerini al
    synsets = wn.synsets(target_word)
    
    if not synsets:
        return None, 0, {}
    
    # 3. Her synset için overlap skorunu hesapla
    best_sense = None
    max_overlap = 0
    overlap_scores = {}
    
    for synset in synsets:
        # Synset tanımını al ve tokenize et
        definition = set(word_tokenize(synset.definition().lower()))
        
        # Ortak kelimeleri bul (overlap)
        overlap = context.intersection(definition)
        overlap_count = len(overlap)
        
        # Skorları kaydet
        overlap_scores[synset] = {
            'overlap': overlap,
            'count': overlap_count,
            'definition': synset.definition()
        }
        
        # En iyi skoru güncelle
        if overlap_count > max_overlap:
            max_overlap = overlap_count
            best_sense = synset
    
    return best_sense, max_overlap, overlap_scores


# 🧪 Test Edelim - "Bank" Örneği

print("🔬 SIMPLE LESK ALGORİTMASI - MANUEL İMPLEMENTASYON")
print("=" * 90)

test_sentences = [
    {"sentence": "I need to go to the bank to deposit money.", "word": "bank", "emoji": "🏦"},
    {"sentence": "The river bank was flooded after the rain.", "word": "bank", "emoji": "🌊"},
    {"sentence": "He keeps coins in his piggy bank.", "word": "bank", "emoji": "🐷"}
]

for i, test in enumerate(test_sentences, 1):
    print(f"\n{'='*90}")
    print(f"{test['emoji']} TEST {i}")
    print(f"{'='*90}")
    print(f"📝 Cümle: \"{test['sentence']}\"")
    print(f"🎯 Hedef Kelime: {test['word'].upper()}")
    
    # Lesk algoritmasını çalıştır
    best_synset, best_score, all_scores = manual_simple_lesk(test['sentence'], test['word'])
    
    print(f"\n🏆 SONUÇ:")
    print(f"   Synset: {best_synset.name()}")
    print(f"   Tanım: {best_synset.definition()}")
    print(f"   Overlap Skoru: {best_score}")
    print(f"   POS: {best_synset.pos()}")
    
    # Tüm synset'lerin skorlarını göster
    print(f"\n📊 TÜM SYNSET SKORLARI:")
    sorted_scores = sorted(all_scores.items(), key=lambda x: x[1]['count'], reverse=True)
    
    for j, (synset, score_info) in enumerate(sorted_scores, 1):
        marker = "🏆" if j == 1 else "  "
        print(f"\n   {marker} {j}. {synset.name()} (Skor: {score_info['count']})")
        print(f"      Tanım: {score_info['definition']}")
        if score_info['overlap']:
            print(f"      Ortak kelimeler: {', '.join(sorted(score_info['overlap']))}")
        else:
            print(f"      Ortak kelimeler: Yok")

print("\n" + "=" * 90)
print("\n✅ Simple Lesk analizi tamamlandı!")
```

```python
# 🚀 Üç Lesk Algoritmasını Karşılaştırma

print("🚀 LESK ALGORİTMALARI KARŞILAŞTIRMASI")
print("=" * 100)
print("📝 Simple Lesk vs Adapted Lesk vs Cosine Lesk\n")

# Test cümleleri
comparison_tests = [
    {"sentence": "I went to the bank to deposit my paycheck.", "word": "bank", "emoji": "🏦"},
    {"sentence": "We sat on the river bank fishing.", "word": "bank", "emoji": "🌊"},
    {"sentence": "The plant needs water to grow.", "word": "plant", "emoji": "🌱"},
    {"sentence": "The factory plant employs 500 workers.", "word": "plant", "emoji": "🏭"},
    {"sentence": "We watched the football match on TV.", "word": "match", "emoji": "⚽"},
    {"sentence": "I need a match to light the candle.", "word": "match", "emoji": "🔥"},
]

# Sonuçları saklamak için
comparison_results = []

for i, test in enumerate(comparison_tests, 1):
    sentence = test['sentence']
    word = test['word']
    emoji = test['emoji']
    
    print(f"\n{'='*100}")
    print(f"{emoji} TEST {i}: '{word.upper()}' kelimesi")
    print(f"{'='*100}")
    print(f"📝 Cümle: \"{sentence}\"\n")
    
    # Tokenize et (PyWSD için gerekli)
    tokens = word_tokenize(sentence)
    
    # 1. Simple Lesk
    try:
        simple_result = simple_lesk(sentence, word)
        simple_synset = simple_result.name() if simple_result else "None"
        simple_def = simple_result.definition() if simple_result else "Bulunamadı"
        print(f"🔬 SIMPLE LESK:")
        print(f"   Synset: {simple_synset}")
        print(f"   Tanım: {simple_def}\n")
    except Exception as e:
        simple_synset = "Error"
        simple_def = str(e)
        print(f"🔬 SIMPLE LESK: Hata - {e}\n")
    
    # 2. Adapted Lesk
    try:
        adapted_result = adapted_lesk(sentence, word)
        adapted_synset = adapted_result.name() if adapted_result else "None"
        adapted_def = adapted_result.definition() if adapted_result else "Bulunamadı"
        print(f"🚀 ADAPTED LESK:")
        print(f"   Synset: {adapted_synset}")
        print(f"   Tanım: {adapted_def}\n")
    except Exception as e:
        adapted_synset = "Error"
        adapted_def = str(e)
        print(f"🚀 ADAPTED LESK: Hata - {e}\n")
    
    # 3. Cosine Lesk
    try:
        cosine_result = cosine_lesk(sentence, word)
        cosine_synset = cosine_result.name() if cosine_result else "None"
        cosine_def = cosine_result.definition() if cosine_result else "Bulunamadı"
        print(f"📐 COSINE LESK:")
        print(f"   Synset: {cosine_synset}")
        print(f"   Tanım: {cosine_def}\n")
    except Exception as e:
        cosine_synset = "Error"
        cosine_def = str(e)
        print(f"📐 COSINE LESK: Hata - {e}\n")
    
    # Sonuçları kaydet
    comparison_results.append({
        'emoji': emoji,
        'word': word,
        'sentence': sentence[:50] + "..." if len(sentence) > 50 else sentence,
        'simple': simple_synset,
        'adapted': adapted_synset,
        'cosine': cosine_synset
    })

# Sonuçları DataFrame'e dönüştür
df_comparison = pd.DataFrame(comparison_results)

print("\n" + "=" * 100)
print("\n📊 KARŞILAŞTIRMA TABLOSU\n")
print(df_comparison.to_string(index=False))

# Algoritmaların anlaşma oranı
print("\n" + "=" * 100)
print("📈 ALGORITMA ANLAŞMA ANALİZİ")
print("=" * 100)

# Tüm algoritmalar aynı sonucu verdi mi?
all_agree = 0
two_agree = 0
none_agree = 0

for _, row in df_comparison.iterrows():
    results = set([row['simple'], row['adapted'], row['cosine']])
    results.discard("Error")
    results.discard("None")
    
    if len(results) == 1:
        all_agree += 1
    elif len(results) == 2:
        two_agree += 1
    else:
        none_agree += 1

total_tests = len(df_comparison)
print(f"\n✅ Tüm algoritmalar anlaştı: {all_agree}/{total_tests} (%{all_agree/total_tests*100:.1f})")
print(f"⚖️  İki algoritma anlaştı: {two_agree}/{total_tests} (%{two_agree/total_tests*100:.1f})")
print(f"❌ Hiçbiri anlaşmadı: {none_agree}/{total_tests} (%{none_agree/total_tests*100:.1f})")
print("=" * 100)
```

```python
# 🎪 İnteraktif WSD Analiz Fonksiyonu

def analyze_word_sense(sentence: str, target_word: str, show_all_synsets: bool = True):
    """
    Verilen cümledeki hedef kelimenin anlamını üç farklı Lesk algoritmasıyla analiz eder
    
    Args:
        sentence (str): Analiz edilecek cümle
        target_word (str): Anlamı bulunacak kelime
        show_all_synsets (bool): Tüm olası synset'leri göster
    """
    print("=" * 100)
    print(f"🎯 KELİME ANLAMI BELİRSİZLİĞİ ANALİZİ")
    print("=" * 100)
    print(f"\n📝 Cümle: \"{sentence}\"")
    print(f"🎯 Hedef Kelime: {target_word.upper()}\n")
    
    # Tüm olası synset'leri göster
    if show_all_synsets:
        all_synsets = wn.synsets(target_word)
        print(f"📚 OLASI ANLAMLAR ({len(all_synsets)} synset bulundu):")
        print("-" * 100)
        for i, synset in enumerate(all_synsets, 1):
            print(f"\n   {i}. {synset.name()}")
            print(f"      📖 Tanım: {synset.definition()}")
            if synset.examples():
                print(f"      💬 Örnek: \"{synset.examples()[0]}\"")
        print("\n" + "=" * 100)
    
    # Üç algoritmayı çalıştır
    print(f"\n🔬 ALGORİTMA SONUÇLARI:\n")
    
    results = {}
    
    # 1. Simple Lesk
    print("1️⃣ SIMPLE LESK:")
    print("-" * 100)
    try:
        simple_result = simple_lesk(sentence, target_word)
        if simple_result:
            results['simple'] = simple_result
            print(f"   ✅ Bulunan Synset: {simple_result.name()}")
            print(f"   📖 Tanım: {simple_result.definition()}")
            print(f"   🏷️  POS: {simple_result.pos()}")
            if simple_result.examples():
                print(f"   💬 Örnek: \"{simple_result.examples()[0]}\"")
        else:
            print("   ❌ Sonuç bulunamadı")
    except Exception as e:
        print(f"   ⚠️ Hata: {e}")
    
    # 2. Adapted Lesk
    print("\n2️⃣ ADAPTED LESK:")
    print("-" * 100)
    try:
        adapted_result = adapted_lesk(sentence, target_word)
        if adapted_result:
            results['adapted'] = adapted_result
            print(f"   ✅ Bulunan Synset: {adapted_result.name()}")
            print(f"   📖 Tanım: {adapted_result.definition()}")
            print(f"   🏷️  POS: {adapted_result.pos()}")
            if adapted_result.examples():
                print(f"   💬 Örnek: \"{adapted_result.examples()[0]}\"")
        else:
            print("   ❌ Sonuç bulunamadı")
    except Exception as e:
        print(f"   ⚠️ Hata: {e}")
    
    # 3. Cosine Lesk
    print("\n3️⃣ COSINE LESK:")
    print("-" * 100)
    try:
        cosine_result = cosine_lesk(sentence, target_word)
        if cosine_result:
            results['cosine'] = cosine_result
            print(f"   ✅ Bulunan Synset: {cosine_result.name()}")
            print(f"   📖 Tanım: {cosine_result.definition()}")
            print(f"   🏷️  POS: {cosine_result.pos()}")
            if cosine_result.examples():
                print(f"   💬 Örnek: \"{cosine_result.examples()[0]}\"")
        else:
            print("   ❌ Sonuç bulunamadı")
    except Exception as e:
        print(f"   ⚠️ Hata: {e}")
    
    # Sonuç özeti
    print("\n" + "=" * 100)
    print("📊 ÖZET:")
    print("=" * 100)
    
    if len(results) == 0:
        print("❌ Hiçbir algoritma sonuç bulamadı")
    elif len(set([r.name() for r in results.values()])) == 1:
        print("✅ TÜM ALGORİTMALAR ANLAŞTI!")
        print(f"🎯 Ortak Sonuç: {list(results.values())[0].name()}")
        print(f"📖 Anlam: {list(results.values())[0].definition()}")
    else:
        print("⚠️ Algoritmalar farklı sonuçlar verdi")
        for algo, synset in results.items():
            print(f"   • {algo.capitalize()}: {synset.name()}")
    
    print("=" * 100)
    return results


# 🧪 Demo Testleri

print("🎪 İNTERAKTİF WSD DEMO\n")

# Test 1
print("\n" + "🔵" * 50 + "\n")
analyze_word_sense(
    sentence="I need to withdraw money from the bank.",
    target_word="bank",
    show_all_synsets=False
)

# Test 2
print("\n" + "🟢" * 50 + "\n")
analyze_word_sense(
    sentence="The plant grows very fast in summer.",
    target_word="plant",
    show_all_synsets=False
)

# Test 3
print("\n" + "🟡" * 50 + "\n")
analyze_word_sense(
    sentence="She plays the guitar beautifully.",
    target_word="play",
    show_all_synsets=False
)
```

## 💼 Best Practices ve İpuçları

### ✅ WSD Kullanırken Dikkat Edilmesi Gerekenler

#### 1. 🎯 Context Seçimi
- Context window'u çok dar tutmayın (minimum ±3 kelime)
- Cümlenin tamamını kullanmak genellikle daha iyi sonuç verir
- Paragraf düzeyinde context bazı durumlarda faydalı olabilir

#### 2. 📊 Algoritma Seçimi
- **Simple Lesk:** Hızlı prototipleme için uygundur
- **Adapted Lesk:** Daha iyi doğruluk istiyorsanız tercih edin
- **Cosine Lesk:** Modern uygulamalar için önerilir

#### 3. 🔍 POS Tagging Kullanımı
- Hedef kelimenin POS etiketini belirlemek doğruluğu artırır
- WordNet'te aramayı sınırlandırarak hızlanma sağlar
- Yanlış POS etiketi sonuçları etkileyebilir

#### 4. ⚠️ Hata Yönetimi
- Her zaman `try-except` blokları kullanın
- Bazı kelimeler WordNet'te olmayabilir
- None dönüş değerini kontrol edin

#### 5. 🚀 Performans Optimizasyonu
- Büyük veri setlerinde batch processing kullanın
- WordNet sorgularını cache'leyin
- Paralel işleme düşünülebilir
