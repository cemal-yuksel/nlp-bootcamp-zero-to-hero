## 🧠 1. Metin Sınıflandırma Nedir?

**Metin Sınıflandırma (Text Classification)**, bir metin belgesini önceden tanımlanmış kategorilerden birine atama işlemidir.

### 🎪 Basit Bir Örnek:

Elinizde bir mesaj var:
```
"Congratulations! You won $1000. Click here to claim your prize!"
```

**Soru:** Bu mesaj **SPAM** mi yoksa **NORMAL (HAM)** mi?

**Cevap:** 🔴 **SPAM!**

İşte metin sınıflandırma tam olarak bunu yapıyor! Bir metni analiz edip kategorisini belirliyor.

    <h3>💡 Önemli Not</h3>
    <p>Metin sınıflandırma, <strong>Supervised Learning (Gözetimli Öğrenme)</strong> kategorisine girer. Yani modeli eğitmek için <strong>etiketlenmiş veri</strong> (labeled data) gerekir!</p>
</div>

## 🔍 2. Kullanım Alanları ve Gerçek Hayat Örnekleri

Metin sınıflandırma günlük hayatımızda birçok alanda kullanılıyor:

### 📧 **1. Spam Tespiti (E-posta Filtreleme)**
- Gmail, Outlook gibi servislerde spam mesajları otomatik olarak filtreleme
- **Örnek:** "Tebrikler! 1 milyon dolar kazandınız" → SPAM

### 😊 **2. Duygu Analizi (Sentiment Analysis)**
- Ürün yorumlarının pozitif/negatif/nötr olarak sınıflandırılması
- **Örnek:** "Bu telefon harika, kesinlikle tavsiye ederim!" → POZİTİF

### 📰 **3. Haber Kategorizasyonu**
- Haberlerin spor, ekonomi, politika gibi kategorilere ayrılması
- **Örnek:** "Galatasaray şampiyonluğa yaklaşıyor" → SPOR

### 🏥 **4. Tıbbi Doküman Sınıflandırma**
- Hasta raporlarının hastalık kategorilerine göre ayrılması
- **Örnek:** "Hasta yüksek ateş ve öksürük şikayeti ile başvurdu" → SOLUNUM SİSTEMİ

### 🎬 **5. İçerik Moderasyonu**
- Sosyal medyada zararlı içeriklerin tespiti
- **Örnek:** Nefret söylemi, şiddet içeren yorumları tespit etme

### 📞 **6. Müşteri Destek Otomasyonu**
- Gelen taleplerin kategorilere ayrılması (fatura, teknik destek, iade vs.)
- **Örnek:** "İnternetim çalışmıyor" → TEKNİK DESTEK

## 💡 4. Temel Kavramlar ve Terimler (Flashcards)

Metin sınıflandırma öğrenirken karşılaşacağımız önemli terimler:

---

<div style="background-color: #e1f5fe; padding: 20px; border-radius: 10px; margin: 10px 0; border: 2px solid #0288d1;">
    <h3>📇 FLASHCARD #1: Corpus (Derlem)</h3>
    <p><strong>Tanım:</strong> İşlem yapılacak tüm metin belgelerinin toplamı</p>
    <p><strong>Örnek:</strong> 5,574 SMS mesajından oluşan veri setimiz bir corpus'tur</p>
</div>

<div style="background-color: #fce4ec; padding: 20px; border-radius: 10px; margin: 10px 0; border: 2px solid #c2185b;">
    <h3>📇 FLASHCARD #2: Token (Jeton)</h3>
    <p><strong>Tanım:</strong> Metnin en küçük anlamlı birimleri (genellikle kelimeler)</p>
    <p><strong>Örnek:</strong> "Hello World" → ["Hello", "World"] (2 token)</p>
</div>

<div style="background-color: #fff9c4; padding: 20px; border-radius: 10px; margin: 10px 0; border: 2px solid #f57f17;">
    <h3>📇 FLASHCARD #3: Stopwords (Durak Kelimeleri)</h3>
    <p><strong>Tanım:</strong> Çok sık kullanılan ama anlam taşımayan kelimeler</p>
    <p><strong>Örnek:</strong> "the", "is", "and", "a", "an" gibi</p>
    <p><strong>Neden kaldırılır?</strong> Model performansını artırmak ve boyut azaltmak için</p>
</div>

<div style="background-color: #f1f8e9; padding: 20px; border-radius: 10px; margin: 10px 0; border: 2px solid #689f38;">
    <h3>📇 FLASHCARD #4: Lemmatization (Kök Bulma)</h3>
    <p><strong>Tanım:</strong> Kelimeleri sözlük kök formlarına (lemma) dönüştürme işlemi</p>
    <p><strong>Örnek:</strong></p>
    <ul>
        <li>"running" → "run"</li>
        <li>"better" → "good"</li>
        <li>"was" → "be"</li>
    </ul>
</div>

<div style="background-color: #e8eaf6; padding: 20px; border-radius: 10px; margin: 10px 0; border: 2px solid #3f51b5;">
    <h3>📇 FLASHCARD #5: Bag of Words (Kelime Torbası)</h3>
    <p><strong>Tanım:</strong> Metni kelime frekanslarına göre sayısal vektöre dönüştürme yöntemi</p>
    <p><strong>Özellik:</strong> Kelime sırasını göz ardı eder, sadece kelime sayılarına bakar</p>
    <p><strong>Örnek:</strong> "cat sat on the mat" → {cat:1, sat:1, on:1, the:1, mat:1}</p>
</div>

<div style="background-color: #fbe9e7; padding: 20px; border-radius: 10px; margin: 10px 0; border: 2px solid #d84315;">
    <h3>📇 FLASHCARD #6: Feature Extraction (Özellik Çıkarımı)</h3>
    <p><strong>Tanım:</strong> Metinleri makine öğrenmesi algoritmaları için sayısal özelliklere dönüştürme</p>
    <p><strong>Yöntemler:</strong></p>
    <ul>
        <li><strong>Count Vectorizer:</strong> Kelime sayısı bazlı</li>
        <li><strong>TF-IDF:</strong> Kelime önemini hesaplayarak</li>
        <li><strong>Word Embeddings:</strong> Word2Vec, GloVe</li>
    </ul>
</div>

<div style="background-color: #e0f2f1; padding: 20px; border-radius: 10px; margin: 10px 0; border: 2px solid #00695c;">
    <h3>📇 FLASHCARD #7: Binary Classification (İkili Sınıflandırma)</h3>
    <p><strong>Tanım:</strong> Verinin iki sınıftan birine atanması</p>
    <p><strong>Örnek:</strong> Spam/Ham, Pozitif/Negatif, Hasta/Sağlıklı</p>
    <p><strong>Projemizde:</strong> SMS mesajlarını SPAM veya HAM olarak sınıflandıracağız</p>
</div>

## 📊 5. Sınıf Dağılımı Analizi

En önemli analiz! Veri setimizde **spam** ve **ham** mesajlarının dağılımı nasıl?

````python
# Sınıf dağılımı
print("="*70)
print("📊 SINIF DAĞILIMI ANALİZİ")
print("="*70)

# Sayısal dağılım
label_counts = df['label'].value_counts()
print(f"\n📈 Mesaj Sayıları:")
print(label_counts)

# Yüzdelik dağılım
label_percentages = df['label'].value_counts(normalize=True) * 100
print(f"\n📊 Yüzdelik Dağılım:")
for label, percentage in label_percentages.items():
    print(f"   {label}: %{percentage:.2f}")

# Oransal analiz
ham_count = label_counts['ham']
spam_count = label_counts['spam']
ratio = ham_count / spam_count

print(f"\n⚖️ Ham/Spam Oranı: {ratio:.2f}:1")
print(f"   (Her {ratio:.0f} normal mesaja karşılık 1 spam mesaj var)")
````

## 📏 6. Metin Uzunluk Analizi

Mesajların uzunluklarını analiz edelim. Spam mesajlar daha uzun mu, yoksa normal mesajlar mı?

````python
# Metin uzunluğu (karakter sayısı) hesaplama
df['text_length'] = df['text'].apply(len)

# Kelime sayısı hesaplama
df['word_count'] = df['text'].apply(lambda x: len(x.split()))

print("="*70)
print("📏 METİN UZUNLUK İSTATİSTİKLERİ")
print("="*70)

# Genel istatistikler
print("\n📊 Tüm Mesajlar İçin:")
print(f"   Ortalama karakter sayısı: {df['text_length'].mean():.2f}")
print(f"   Ortalama kelime sayısı: {df['word_count'].mean():.2f}")

# Sınıflara göre istatistikler
print("\n📊 Sınıflara Göre İstatistikler:")
print("-" * 70)

for label in ['ham', 'spam']:
    subset = df[df['label'] == label]
    print(f"\n🏷️ {label.upper()} Mesajları:")
    print(f"   Ortalama karakter: {subset['text_length'].mean():.2f}")
    print(f"   Ortalama kelime: {subset['word_count'].mean():.2f}")
    print(f"   En kısa mesaj: {subset['text_length'].min()} karakter")
    print(f"   En uzun mesaj: {subset['text_length'].max()} karakter")
````

````python
# Görselleştirme: Histogram ve Boxplot
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Karakter Sayısı - Histogram
axes[0, 0].hist(df[df['label'] == 'ham']['text_length'], bins=50, alpha=0.7, 
                label='Ham', color='#4CAF50', edgecolor='black')
axes[0, 0].hist(df[df['label'] == 'spam']['text_length'], bins=50, alpha=0.7, 
                label='Spam', color='#F44336', edgecolor='black')
axes[0, 0].set_xlabel('Karakter Sayısı', fontsize=11, weight='bold')
axes[0, 0].set_ylabel('Frekans', fontsize=11, weight='bold')
axes[0, 0].set_title('📏 Karakter Sayısı Dağılımı', fontsize=13, weight='bold', pad=15)
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# 2. Kelime Sayısı - Histogram
axes[0, 1].hist(df[df['label'] == 'ham']['word_count'], bins=50, alpha=0.7, 
                label='Ham', color='#4CAF50', edgecolor='black')
axes[0, 1].hist(df[df['label'] == 'spam']['word_count'], bins=50, alpha=0.7, 
                label='Spam', color='#F44336', edgecolor='black')
axes[0, 1].set_xlabel('Kelime Sayısı', fontsize=11, weight='bold')
axes[0, 1].set_ylabel('Frekans', fontsize=11, weight='bold')
axes[0, 1].set_title('📝 Kelime Sayısı Dağılımı', fontsize=13, weight='bold', pad=15)
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

# 3. Karakter Sayısı - Boxplot
bp1 = axes[1, 0].boxplot([df[df['label'] == 'ham']['text_length'], 
                           df[df['label'] == 'spam']['text_length']], 
                          labels=['Ham', 'Spam'], patch_artist=True,
                          boxprops=dict(facecolor='lightblue', alpha=0.7),
                          medianprops=dict(color='red', linewidth=2))
axes[1, 0].set_ylabel('Karakter Sayısı', fontsize=11, weight='bold')
axes[1, 0].set_title('📊 Karakter Sayısı Boxplot', fontsize=13, weight='bold', pad=15)
axes[1, 0].grid(axis='y', alpha=0.3)

# 4. Kelime Sayısı - Boxplot
bp2 = axes[1, 1].boxplot([df[df['label'] == 'ham']['word_count'], 
                           df[df['label'] == 'spam']['word_count']], 
                          labels=['Ham', 'Spam'], patch_artist=True,
                          boxprops=dict(facecolor='lightcoral', alpha=0.7),
                          medianprops=dict(color='darkblue', linewidth=2))
axes[1, 1].set_ylabel('Kelime Sayısı', fontsize=11, weight='bold')
axes[1, 1].set_title('📊 Kelime Sayısı Boxplot', fontsize=13, weight='bold', pad=15)
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✅ Görselleştirme tamamlandı!")
````

## 🔬 3. Tek Bir Örnek Üzerinde Preprocessing Adımlarını Görelim

Önce bir örnek mesajı adım adım işleyerek her aşamayı anlayalım:

````python
# Örnek bir spam mesajı seçelim
sample_text = "FREE FREE! Win £1000 cash prize NOW!!! Call 09061701461 to claim..."

print("="*70)
print("🔬 ADIM ADIM METİN ÖN İŞLEME")
print("="*70)

print("\n📝 ORIJINAL METİN:")
print(f"   '{sample_text}'")
print(f"   Uzunluk: {len(sample_text)} karakter")

# ADIM 1: Özel karakterleri kaldır (sadece harf bırak)
# [^a-zA-Z] = harf olmayanları bul ve boşlukla değiştir
step1 = re.sub("[^a-zA-Z]", " ", sample_text)
print("\n🧹 ADIM 1 - Özel Karakterler Kaldırıldı:")
print(f"   '{step1}'")

# ADIM 2: Lowercase (küçük harfe çevir)
step2 = step1.lower()
print("\n🔤 ADIM 2 - Lowercase Dönüşümü:")
print(f"   '{step2}'")

# ADIM 3: Tokenization (kelimelere ayır)
step3 = step2.split()
print("\n✂️ ADIM 3 - Tokenization:")
print(f"   {step3}")
print(f"   Kelime sayısı: {len(step3)}")

# ADIM 4: Stopwords kaldır
step4 = [word for word in step3 if word not in stopwords.words('english')]
print("\n🚫 ADIM 4 - Stopwords Kaldırıldı:")
print(f"   {step4}")
print(f"   Kalan kelime sayısı: {len(step4)}")

# ADIM 5: Lemmatization (kök bulma)
lemmatizer = WordNetLemmatizer()
step5 = [lemmatizer.lemmatize(word) for word in step4]
print("\n🌱 ADIM 5 - Lemmatization:")
print(f"   {step5}")

# Final metin
final_text = " ".join(step5)
print("\n✅ FİNAL METİN:")
print(f"   '{final_text}'")
print(f"   Final uzunluk: {len(final_text)} karakter")

print("\n" + "="*70)
print(f"📊 ÖZET: {len(sample_text)} karakter → {len(final_text)} karakter")
print(f"         {len(sample_text.split())} kelime → {len(step5)} kelime")
print(f"         %{((len(sample_text) - len(final_text))/len(sample_text) * 100):.1f} boyut azaltma!")
print("="*70)
````

---

## 🔄 4. Tüm Veri Setine Preprocessing Uygulama

Şimdi öğrendiğimiz tüm adımları bir fonksiyon haline getirip tüm veri setine uygulayalım:

````python
# Preprocessing fonksiyonu oluşturma
def preprocess_text(text):
    """
    Bir metni ön işleme adımlarından geçirir.
    
    Parametreler:
    -----------
    text : str
        İşlenecek ham metin
        
    Döndürür:
    ---------
    str
        İşlenmiş (temizlenmiş) metin
    """
    # 1. Özel karakterleri kaldır (sadece harfler kalsın)
    text = re.sub("[^a-zA-Z]", " ", text)
    
    # 2. Küçük harfe çevir
    text = text.lower()
    
    # 3. Kelimelere ayır (tokenization)
    words = text.split()
    
    # 4. Stopwords'leri kaldır
    words = [word for word in words if word not in stopwords.words('english')]
    
    # 5. Lemmatization (kök bulma)
    lemmatizer = WordNetLemmatizer()
    words = [lemmatizer.lemmatize(word) for word in words]
    
    # 6. Kelimeleri birleştir
    cleaned_text = " ".join(words)
    
    return cleaned_text

# Test edelim
test_text = "I'm LOVING this product! It's AMAZING!!!"
print(f"Orijinal: {test_text}")
print(f"İşlenmiş: {preprocess_text(test_text)}")
print("\n✅ Fonksiyon hazır!")
````

````python
# Tüm veri setine preprocessing uygulama
print("="*70)
print("🔄 TÜM VERİ SETİNE PREPROCESSING UYGULANYOR...")
print("="*70)

import time
start_time = time.time()

# Her bir mesajı preprocessing fonksiyonundan geçir
corpus = []
text_list = df['text'].tolist()

for i, text in enumerate(text_list):
    cleaned_text = preprocess_text(text)
    corpus.append(cleaned_text)
    
    # Her 1000 mesajda bir ilerleme göster
    if (i + 1) % 1000 == 0:
        print(f"   ✓ {i + 1}/{len(text_list)} mesaj işlendi...")

# Yeni sütun olarak ekle
df['cleaned_text'] = corpus

end_time = time.time()
elapsed_time = end_time - start_time

print("\n✅ Preprocessing tamamlandı!")
print(f"⏱️ Süre: {elapsed_time:.2f} saniye")
print(f"📊 İşlenen mesaj sayısı: {len(corpus)}")

# Sonuçları göster
print("\n" + "="*70)
print("📋 ÖNCESİ ve SONRASI KARŞILAŞTIRMA")
print("="*70)
print("\nİlk 3 mesaj örneği:\n")

for i in range(3):
    print(f"🔴 Orijinal {i+1}:")
    print(f"   {df['text'].iloc[i][:100]}...")
    print(f"\n🟢 İşlenmiş {i+1}:")
    print(f"   {df['cleaned_text'].iloc[i][:100]}...")
    print("\n" + "-"*70 + "\n")
````

## ⚖️ 5. SMOTE ile Veri Dengeleme (Oversampling)

Veri setimizde **imbalanced (dengesiz)** problem var (Ham: %87, Spam: %13). Bu sorunu **SMOTE** ile çözeceğiz!

<div style="background-color: #e3f2fd; padding: 20px; border-radius: 10px; margin: 10px 0; border: 2px solid #0288d1;">
    <h3>📇 SMOTE Nedir?</h3>
    <p><strong>SMOTE (Synthetic Minority Over-sampling Technique):</strong> Azınlık sınıftan (spam) sentetik örnekler üreterek veri setini dengeler.</p>
    <p><strong>Nasıl Çalışır?</strong></p>
    <ul>
        <li>Azınlık sınıftan (spam) örnekler seçilir</li>
        <li>Bu örneklerin komşuları bulunur</li>
        <li>Komşular arasında <strong>sentetik yeni örnekler</strong> oluşturulur</li>
        <li>Böylece veri seti dengelenir</li>
    </ul>
    <p><strong>⚠️ Önemli:</strong> SMOTE'u sadece <strong>eğitim setine</strong> uygularız, test setine asla!</p>
</div>

### 📦 SMOTE Kütüphanesini Kuralım

Öncelikle `imbalanced-learn` kütüphanesini kurmalıyız:

````python
# SMOTE için gerekli kütüphaneleri import edelim
try:
    from imblearn.over_sampling import SMOTE
    print("✅ imbalanced-learn kütüphanesi zaten yüklü!")
except ImportError:
    print("📦 imbalanced-learn kütüphanesi kuruluyor...")
    import sys
    !{sys.executable} -m pip install imbalanced-learn -q
    from imblearn.over_sampling import SMOTE
    print("✅ imbalanced-learn başarıyla kuruldu!")

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

print("✅ Tüm kütüphaneler hazır!")
````

### 🔀 Train-Test Split (Önce Ayırma İşlemi)

````python
# X (özellikler) ve y (hedef değişken) ayırma
X = df['cleaned_text']  # İşlenmiş metinler
y = df['label']          # Etiketler (spam/ham)

# Train-Test Split (%80 eğitim, %20 test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20,      # Test için %20
    random_state=42,     # Tekrarlanabilirlik için sabit seed
    stratify=y           # Sınıf oranlarını koruyarak böl
)

print("="*70)
print("🔀 TRAİN-TEST SPLIT TAMAMLANDI")
print("="*70)

print(f"\n📊 Eğitim seti boyutu: {len(X_train)} mesaj")
print(f"📊 Test seti boyutu: {len(X_test)} mesaj")

print(f"\n📈 Eğitim seti sınıf dağılımı:")
print(y_train.value_counts())
print(f"\n📈 Test seti sınıf dağılımı:")
print(y_test.value_counts())
````

### 🔢 Bag of Words (Feature Extraction)

````python
# CountVectorizer ile metinleri sayısal vektörlere dönüştürme
cv = CountVectorizer(max_features=3000)  # En sık kullanılan 3000 kelimeyi al

# Eğitim seti üzerinde fit ve transform
X_train_cv = cv.fit_transform(X_train)

# Test seti üzerinde sadece transform (fit etmiyoruz!)
X_test_cv = cv.transform(X_test)

print("="*70)
print("🔢 BAG OF WORDS (CountVectorizer) UYGULANDIĞI")
print("="*70)

print(f"\n📊 Eğitim seti özellik matrisi boyutu: {X_train_cv.shape}")
print(f"   ({X_train_cv.shape[0]} mesaj x {X_train_cv.shape[1]} özellik)")

print(f"\n📊 Test seti özellik matrisi boyutu: {X_test_cv.shape}")
print(f"   ({X_test_cv.shape[0]} mesaj x {X_test_cv.shape[1]} özellik)")

print(f"\n📝 Vocabulary (Sözlük) boyutu: {len(cv.vocabulary_)}")
print(f"📝 İlk 20 kelime: {list(cv.vocabulary_.keys())[:20]}")
````

### ⚖️ SMOTE Uygulama (Sadece Eğitim Setine!)

````python
print("="*70)
print("⚖️ SMOTE İLE VERİ DENGELEME")
print("="*70)

# SMOTE öncesi sınıf dağılımı
print("\n📊 SMOTE ÖNCESİ Eğitim Seti Sınıf Dağılımı:")
print(y_train.value_counts())
print(f"\nToplam: {len(y_train)} mesaj")

# SMOTE uygulama
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train_cv, y_train)

# SMOTE sonrası sınıf dağılımı
print("\n" + "="*70)
print("📊 SMOTE SONRASI Eğitim Seti Sınıf Dağılımı:")
print(pd.Series(y_train_balanced).value_counts())
print(f"\nToplam: {len(y_train_balanced)} mesaj")

print("\n✅ Veri seti başarıyla dengelendi!")
print(f"📈 Spam mesaj sayısı: {(y_train=='spam').sum()} → {(y_train_balanced=='spam').sum()}")
print(f"📈 Ham mesaj sayısı: {(y_train=='ham').sum()} → {(y_train_balanced=='ham').sum()}")
````

````python
# Performans metrikleri hesapla
dt_accuracy = accuracy_score(y_test, dt_predictions)
dt_precision = precision_score(y_test, dt_predictions, pos_label='spam')
dt_recall = recall_score(y_test, dt_predictions, pos_label='spam')
dt_f1 = f1_score(y_test, dt_predictions, pos_label='spam')

print("\n" + "="*70)
print("📊 DECISION TREE PERFORMANS METRİKLERİ")
print("="*70)

print(f"\n🎯 Accuracy (Doğruluk):   %{dt_accuracy*100:.2f}")
print(f"🎯 Precision (Kesinlik):  %{dt_precision*100:.2f}")
print(f"🎯 Recall (Duyarlılık):   %{dt_recall*100:.2f}")
print(f"🎯 F1-Score:              %{dt_f1*100:.2f}")

print("\n" + "="*70)
print("📋 DETAYLI SINIFLANDIRMA RAPORU")
print("="*70)
print(classification_report(y_test, dt_predictions, target_names=['ham', 'spam']))
````

## 🌳 4. MODEL 3: Random Forest Classifier

**Random Forest**, çoklu karar ağaçlarının bir araya gelmesiyle oluşan güçlü bir ensemble yöntemidir!

<div style="background-color: #e8f5e9; padding: 20px; border-radius: 10px; margin: 10px 0; border: 2px solid #4caf50;">
    <h3>🌳 Random Forest Nasıl Çalışır?</h3>
    <p><strong>Mantık:</strong> Birden fazla karar ağacı oluşturup, çoğunluk oylamasıyla karar verir</p>
    <p><strong>Avantajları:</strong></p>
    <ul>
        <li>Tek bir karar ağacından çok daha güçlü</li>
        <li>Overfitting'e karşı dayanıklı</li>

    <p><strong>Dezavantajları:</strong></p>
    <ul>
        <li>Eğitim süresi daha uzun</li>

        <li>Yorumlanması zordur</li>

````python
print("="*70)
print("🌳 RANDOM FOREST CLASSIFIER EĞİTİMİ")
print("="*70)

# Model oluştur
rf_model = RandomForestClassifier(
    n_estimators=100,     # Ağaç sayısı
    max_depth=10,         # Maksimum derinlik
    random_state=42,
    n_jobs=-1             # Tüm CPU çekirdeklerini kullan
)

# Eğitim süresi ölç
start_time = time.time()

# Modeli eğit
rf_model.fit(X_train_balanced, y_train_balanced)

# Eğitim süresi
rf_train_time = time.time() - start_time

print(f"\n✅ Model eğitimi tamamlandı!")
print(f"⏱️ Eğitim süresi: {rf_train_time:.3f} saniye")
print(f"🌳 Kullanılan ağaç sayısı: {rf_model.n_estimators}")

# Test seti üzerinde tahmin yap
start_time = time.time()
rf_predictions = rf_model.predict(X_test_cv)
rf_predict_time = time.time() - start_time

print(f"⏱️ Tahmin süresi: {rf_predict_time:.3f} saniye")
````

````python
# Performans metrikleri
rf_accuracy = accuracy_score(y_test, rf_predictions)
rf_precision = precision_score(y_test, rf_predictions, pos_label='spam')
rf_recall = recall_score(y_test, rf_predictions, pos_label='spam')
rf_f1 = f1_score(y_test, rf_predictions, pos_label='spam')

print("\n" + "="*70)
print("📊 RANDOM FOREST PERFORMANS METRİKLERİ")
print("="*70)

print(f"\n🎯 Accuracy (Doğruluk):   %{rf_accuracy*100:.2f}")
print(f"🎯 Precision (Kesinlik):  %{rf_precision*100:.2f}")
print(f"🎯 Recall (Duyarlılık):   %{rf_recall*100:.2f}")
print(f"🎯 F1-Score:              %{rf_f1*100:.2f}")

print("\n" + "="*70)
print("📋 DETAYLI SINIFLANDIRMA RAPORU")
print("="*70)
print(classification_report(y_test, rf_predictions, target_names=['ham', 'spam']))
````
