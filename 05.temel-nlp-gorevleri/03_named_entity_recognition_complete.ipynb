## 🤔 **Named Entity Recognition (NER) Nedir?**

<div style="background-color: #e3f2fd; padding: 20px; border-left: 5px solid #2196F3; border-radius: 5px; margin: 15px 0;">
    <h3 style="color: #1976D2; margin-top: 0;">📖 Tanım</h3>
    <p style="font-size: 1.1em; line-height: 1.6;">
        <strong>Named Entity Recognition (NER)</strong>, bir metin içerisindeki <strong>özel isimleri</strong> (kişi, kurum, yer, tarih, para birimi vb.) otomatik olarak tanımlayan ve sınıflandıran bir <strong>Doğal Dil İşleme (NLP)</strong> görevidir.
    </p>
</div>

### 💡 **Basit Bir Örnek:**

```
"Ahmet, 15 Mart 2024 tarihinde İstanbul'da Microsoft şirketinde çalışmaya başladı."
```

**NER Sonucu:**
- 👤 **Ahmet** → PERSON (Kişi)
- 📅 **15 Mart 2024** → DATE (Tarih)
- 🌍 **İstanbul** → GPE (Coğrafi Konum)
- 🏢 **Microsoft** → ORG (Organizasyon)

## 🎨 **NER Nasıl Görselleştirilir?**

<div style="background-color: #fff3e0; padding: 20px; border-radius: 10px; border-left: 5px solid #FF9800; margin: 20px 0;">
    <h3 style="color: #E65100; margin-top: 0;">🖼️ displaCy ile Görselleştirme</h3>
    <p style="line-height: 1.6;">
        spaCy'nin <strong>displaCy</strong> modülü, NER sonuçlarını renkli ve interaktif bir şekilde görselleştirir. Her varlık türü farklı bir renkle vurgulanır.
    </p>
</div>

### 📸 **Örnek Görselleştirme:**

```python
from spacy import displacy

text = "Apple CEO Tim Cook, 15 Mart 2024'te Kaliforniya'da yeni ürünleri tanıttı."
doc = nlp(text)
displacy.render(doc, style="ent", jupyter=True)
```

**Çıktı:** Metin içinde her varlık renkli kutular içinde vurgulanır:
- 🟦 <span style="background-color: #7aecec; padding: 2px 6px; border-radius: 3px;">Apple → ORG</span>

````python
# 🌍 spaCy İngilizce dil modelini indirelim
# Bu işlem birkaç dakika sürebilir (model boyutu: ~12 MB)

import subprocess
import sys

def download_spacy_model(model_name):
    """
    spaCy dil modelini indirir
    
    Args:
        model_name (str): İndirilecek model adı (örn: 'en_core_web_sm')
    """
    try:
        print(f"🌍 {model_name} modeli indiriliyor...")
        print("⏳ Lütfen bekleyin, bu işlem 1-2 dakika sürebilir...\n")
        
        # spaCy modelini indir
        subprocess.check_call([
            sys.executable, 
            "-m", 
            "spacy", 
            "download", 
            model_name
        ])
        
        print(f"\n✅ {model_name} başarıyla indirildi!")
        return True
        
    except subprocess.CalledProcessError as e:
        print(f"\n❌ Model indirme hatası: {e}")
        return False

# İngilizce small modelini indir
success = download_spacy_model("en_core_web_sm")

if success:
    print("\n🎉 Harika! Artık NER işlemlerine başlayabilirsiniz!")
````

    <p style="margin: 0;">Bir <strong>entity</strong> tek bir token olabilir ("Microsoft") veya birden fazla token içerebilir ("Steve Jobs", "New York City").</p>
</div>

````python
# 🔤 Token vs Entity farkını anlamak

import spacy

# Model yükle
nlp = spacy.load("en_core_web_sm")

# Örnek metin
sample_text = "Apple CEO Tim Cook announced iPhone 15 in Cupertino on September 12, 2023."

# Metni işle
doc = nlp(sample_text)

print("=" * 70)
print("🔍 TOKEN ANALİZİ")
print("=" * 70)
print(f"\n📄 Metin: '{sample_text}'\n")

# Token'ları göster
print("📌 Token'lar (Kelimeler):")
print("-" * 70)
for i, token in enumerate(doc, 1):
    print(f"{i:2d}. '{token.text}' (pos: {token.idx})")

print(f"\n✅ Toplam Token Sayısı: {len(doc)}")

# Entity'leri göster
print("\n" + "=" * 70)
print("🎯 ENTITY (VARLIK) ANALİZİ")
print("=" * 70 + "\n")

print("📌 Tespit Edilen Varlıklar:")
print("-" * 70)

if doc.ents:
    for i, ent in enumerate(doc.ents, 1):
        # Entity'nin kaç token içerdiğini hesapla
        token_count = len(ent)
        
        # Emoji seç
        emoji_map = {
            'PERSON': '👤', 'ORG': '🏢', 'GPE': '🌍',
            'DATE': '📅', 'PRODUCT': '📦', 'TIME': '⏰'
        }
        emoji = emoji_map.get(ent.label_, '🏷️')
        
        print(f"{i}. {emoji} '{ent.text}'")
        print(f"   ├─ Tür: {ent.label_}")
        print(f"   ├─ Token Sayısı: {token_count}")
        print(f"   └─ Pozisyon: {ent.start_char}-{ent.end_char}\n")
    
    print(f"✅ Toplam Entity Sayısı: {len(doc.ents)}")
else:
    print("❌ Hiç varlık bulunamadı.")

print("\n" + "=" * 70)
````

````python
# 📊 NER sonuçlarını pandas DataFrame'e çevirme

import pandas as pd
import spacy

# Model yükle
nlp = spacy.load("en_core_web_sm")

# Zengin içerikli örnek metin
text = """
Amazon CEO Jeff Bezos announced a $10 billion investment in climate initiatives.
The announcement was made in Seattle, Washington on March 15, 2024.
Microsoft and Google also pledged support, with Bill Gates praising the initiative.
The project will span across 50 countries including United States, China, and India.
"""

# Metni işle
doc = nlp(text)

print("📄 Analiz Edilen Metin:")
print("=" * 70)
print(text.strip())
print("=" * 70)

# Entity bilgilerini liste olarak topla
entities_data = []

for ent in doc.ents:
    entities_data.append({
        'Varlık': ent.text,              # Varlık metni
        'Tür': ent.label_,               # Varlık türü
        'Açıklama': spacy.explain(ent.label_),  # Türün açıklaması
        'Başlangıç': ent.start_char,     # Başlangıç pozisyonu
        'Bitiş': ent.end_char,           # Bitiş pozisyonu
        'Uzunluk': len(ent.text)         # Karakter uzunluğu
    })

# DataFrame oluştur
df = pd.DataFrame(entities_data)

print("\n📊 VARLIK TABLOSU:")
print("=" * 70)

if not df.empty:
    # DataFrame'i güzel bir formatta göster
    print(df.to_string(index=False))
    
    print("\n" + "=" * 70)
    print(f"✅ Toplam {len(df)} varlık tespit edildi!")
    
    # Özet istatistikler
    print("\n📈 ÖZET İSTATİSTİKLER:")
    print("-" * 70)
    entity_counts = df['Tür'].value_counts()
    for entity_type, count in entity_counts.items():
        print(f"  • {entity_type}: {count} adet")
else:
    print("❌ Hiç varlık bulunamadı.")

# DataFrame'i değişkene kaydet (sonraki analizler için)
entities_df = df
````

### 💡 **Örnek Metinler:**

```python
# Örnek 1: Haber metni
"President Joe Biden met with Prime Minister of Japan in Washington DC yesterday."

# Örnek 2: Teknoloji haberi  
"Elon Musk's Tesla announced a new Cybertruck model priced at $79,999."

# Örnek 3: Spor haberi
"Lionel Messi scored 3 goals for Inter Miami against LA Galaxy on Saturday."
```

````python
# 🎮 Kendi metninizi buraya yazın ve NER'i test edin!

import spacy
from spacy import displacy

# Model yükle
nlp = spacy.load("en_core_web_sm")

# 👇 BU KISMI DEĞİŞTİRİN - Kendi metninizi yazın
your_text = """
Elon Musk's Tesla announced a new Cybertruck model priced at $79,999 
at the Tesla Gigafactory in Austin, Texas on October 1st, 2023.
The event was attended by over 5,000 people.
"""

print("🎯 SİZİN METNİNİZ:")
print("=" * 70)
print(your_text.strip())
print("=" * 70)

# Metni işle
doc = nlp(your_text)

# HTML görselleştirme
print("\n🎨 GÖRSEL ANALİZ:\n")
try:
    from IPython.display import display, HTML
    html = displacy.render(doc, style="ent", jupyter=False)
    display(HTML(html))
except:
    # Basit metin formatı
    for ent in doc.ents:
        print(f"[{ent.label_}] {ent.text}")

# Detaylı sonuçlar
print("\n📊 DETAYLI SONUÇLAR:")
print("=" * 70 + "\n")

if doc.ents:
    # Emoji haritası
    emoji_map = {
        'PERSON': '👤', 'ORG': '🏢', 'GPE': '🌍', 'LOC': '📍',
        'DATE': '📅', 'TIME': '⏰', 'MONEY': '💰', 'PERCENT': '📊',
        'PRODUCT': '📦', 'EVENT': '🎭', 'QUANTITY': '🔢', 'CARDINAL': '🔢'
    }
    
    # Varlık türlerine göre grupla
    entity_groups = {}
    for ent in doc.ents:
        if ent.label_ not in entity_groups:
            entity_groups[ent.label_] = []
        entity_groups[ent.label_].append(ent.text)
    
    # Her grup için sonuçları göster
    for label, entities in sorted(entity_groups.items()):
        emoji = emoji_map.get(label, '🏷️')
        label_desc = spacy.explain(label) or label
        print(f"{emoji} {label} ({label_desc}):")
        for entity in entities:
            print(f"   • {entity}")
        print()
    
    print("=" * 70)
    print(f"✅ Toplam {len(doc.ents)} varlık tespit edildi!")
    print(f"📈 {len(entity_groups)} farklı varlık türü bulundu.")
    print("=" * 70)
else:
    print("❌ Hiç varlık bulunamadı.")
    print("💡 İpucu: Metinde kişi, kurum, yer veya tarih isimleri ekleyin.")
    print("=" * 70)

print("\n💡 Yukarıdaki 'your_text' değişkenini değiştirerek farklı metinler deneyin!")
````

## ⚡ Batch NER İşleme

Şimdi tüm haber metinleri üzerinde toplu NER analizi yapacağız. Bu, gerçek dünya uygulamalarında **çok yaygın bir senaryodur**.

<div style="background: linear-gradient(135deg, #ffecd2 0%, #fcb69f 100%); padding: 20px; border-radius: 10px; margin: 15px 0;">
    <h3 style="color: #d63031; margin-top: 0;">🎯 Batch İşleme Stratejisi</h3>
    <p style="margin: 0; line-height: 1.8;">
        <strong>nlp.pipe()</strong> metodunu kullanarak birden fazla metni <strong>verimli bir şekilde</strong> işleyeceğiz. 
        Bu yöntem, her metni ayrı ayrı işlemekten çok daha hızlıdır çünkü <strong>batch processing</strong> yapar.
    </p>
</div>

````python
# ⚡ Tüm haberleri batch olarak işleyelim ve entity'leri çıkaralım
import time

print("=" * 80)
print("⚡ BATCH NER ANALİZİ BAŞLIYOR...")
print("=" * 80)

# Zamanı ölçelim
start_time = time.time()

# Tüm haber metinlerini alalım
texts = news_df['text'].tolist()

# nlp.pipe() ile batch processing yapalım (çok daha verimli!)
docs = list(nlp.pipe(texts))

# Her haber için entity'leri toplayalım
all_entities = []

for idx, doc in enumerate(docs):
    news_id = news_df.iloc[idx]['id']
    category = news_df.iloc[idx]['category']
    title = news_df.iloc[idx]['title']
    
    print(f"\n📰 Haber #{news_id} - [{category}] {title}")
    print(f"   Bulunan entity sayısı: {len(doc.ents)}")
    
    # Her entity'yi kaydet
    for ent in doc.ents:
        all_entities.append({
            'news_id': news_id,
            'category': category,
            'title': title,
            'entity_text': ent.text,
            'entity_type': ent.label_,
            'entity_type_desc': spacy.explain(ent.label_),
            'start_char': ent.start_char,
            'end_char': ent.end_char
        })
    
    # İlk 3 entity'yi göster
    if doc.ents:
        print("   🔹 Öne çıkan entity'ler:")
        for ent in list(doc.ents)[:3]:
            emoji = {'PERSON': '👤', 'ORG': '🏢', 'GPE': '🌍', 
                    'DATE': '📅', 'MONEY': '💰', 'CARDINAL': '🔢'}.get(ent.label_, '📌')
            print(f"      {emoji} {ent.text} ({ent.label_})")

# İşlem süresini hesapla
end_time = time.time()
processing_time = end_time - start_time

# Genel sonuçları göster
print("\n" + "=" * 80)
print("✅ BATCH İŞLEME TAMAMLANDI!")
print("=" * 80)
print(f"⏱️  İşlem süresi: {processing_time:.2f} saniye")
print(f"📊 Toplam işlenen haber: {len(docs)}")
print(f"🎯 Toplam bulunan entity: {len(all_entities)}")
print(f"⚡ Ortalama işlem hızı: {len(docs)/processing_time:.2f} haber/saniye")

# DataFrame'e çevirelim
entities_full_df = pd.DataFrame(all_entities)
print(f"\n✅ Entity DataFrame oluşturuldu: {entities_full_df.shape[0]} satır, {entities_full_df.shape[1]} sütun")
````

Entity'lerin **karakter uzunluklarını** analiz edelim. Bu, hangi tür entity'lerin daha uzun olduğunu anlamamızı sağlar.

````python
# 📏 Entity uzunluk analizi
# Her entity'nin karakter uzunluğunu hesaplayalım
entities_full_df['entity_length'] = entities_full_df['entity_text'].apply(len)

print("=" * 80)
print("📏 ENTITY UZUNLUK İSTATİSTİKLERİ")
print("=" * 80)

# Genel istatistikler
print(f"\nGenel İstatistikler:")
print(f"  📊 Ortalama uzunluk: {entities_full_df['entity_length'].mean():.2f} karakter")
print(f"  📏 En kısa entity: {entities_full_df['entity_length'].min()} karakter")
print(f"  📏 En uzun entity: {entities_full_df['entity_length'].max()} karakter")
print(f"  📊 Medyan uzunluk: {entities_full_df['entity_length'].median():.2f} karakter")

# Entity türüne göre ortalama uzunluk
print("\n" + "=" * 80)
print("📊 ENTITY TÜRÜNE GÖRE ORTALAMA UZUNLUK")
print("=" * 80)

avg_length_by_type = entities_full_df.groupby('entity_type')['entity_length'].agg(['mean', 'min', 'max', 'count'])
avg_length_by_type = avg_length_by_type.sort_values('mean', ascending=False)

for entity_type, row in avg_length_by_type.iterrows():
    emoji = entity_emojis.get(entity_type, '📌')
    print(f"{emoji} {entity_type:12s} | Ort: {row['mean']:5.1f} | Min: {int(row['min']):2d} | Max: {int(row['max']):2d} | Sayı: {int(row['count']):3d}")

# En uzun entity'leri bulalım
print("\n" + "=" * 80)
print("🏆 EN UZUN ENTITY'LER (İLK 10)")
print("=" * 80)

longest_entities = entities_full_df.nlargest(10, 'entity_length')[['entity_text', 'entity_type', 'entity_length', 'category']]

for idx, (_, row) in enumerate(longest_entities.iterrows(), 1):
    emoji = entity_emojis.get(row['entity_type'], '📌')
    print(f"{idx:2d}. {emoji} {row['entity_text']:40s} | {row['entity_length']:2d} karakter | [{row['entity_type']}] | {row['category']}")

# Görselleştirme - Box plot
fig, ax = plt.subplots(figsize=(14, 6))

# Entity türlerine göre uzunluk dağılımı (box plot)
entity_types_for_plot = entities_full_df['entity_type'].value_counts().head(8).index
filtered_df = entities_full_df[entities_full_df['entity_type'].isin(entity_types_for_plot)]

box_plot_data = [filtered_df[filtered_df['entity_type'] == et]['entity_length'].values 
                 for et in entity_types_for_plot]

bp = ax.boxplot(box_plot_data, labels=entity_types_for_plot, patch_artist=True)

# Renklendirme
colors = plt.cm.Pastel1(range(len(bp['boxes'])))
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)

ax.set_xlabel('Entity Türü', fontsize=12, fontweight='bold')
ax.set_ylabel('Karakter Uzunluğu', fontsize=12, fontweight='bold')
ax.set_title('📏 Entity Türlerine Göre Uzunluk Dağılımı (Box Plot)', fontsize=14, fontweight='bold', pad=20)
ax.tick_params(axis='x', rotation=45)
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✅ Uzunluk analizi tamamlandı!")
````

## 🔗 4. Entity Co-occurrence Analizi

    bunlar arasında bir ilişki var demektir. Bu tür analizler, bilgi çıkarımı ve ağ görselleştirmeleri için kullanılır.
</div>

````python
# 🔗 Entity Co-occurrence (Birlikte Geçme) Analizi
from collections import defaultdict
from itertools import combinations

print("=" * 80)
print("🔗 ENTITY CO-OCCURRENCE ANALİZİ")
print("=" * 80)

# Her haber için entity çiftlerini bulalım
co_occurrence = defaultdict(int)

for news_id in entities_full_df['news_id'].unique():
    # Bu haberdeki tüm entity'leri al
    news_entities = entities_full_df[entities_full_df['news_id'] == news_id]['entity_text'].unique()
    
    # Entity çiftlerini oluştur (kombinasyonlar)
    for entity1, entity2 in combinations(sorted(news_entities), 2):
        pair = (entity1, entity2)
        co_occurrence[pair] += 1

# En çok birlikte geçen entity çiftlerini bulalım
top_pairs = sorted(co_occurrence.items(), key=lambda x: x[1], reverse=True)[:15]

print(f"\n🔝 EN ÇOK BİRLİKTE GEÇEN ENTITY ÇİFTLERİ (İLK 15)\n")

for idx, ((entity1, entity2), count) in enumerate(top_pairs, 1):
    # Entity türlerini bulalım
    type1 = entities_full_df[entities_full_df['entity_text'] == entity1]['entity_type'].iloc[0]
    type2 = entities_full_df[entities_full_df['entity_text'] == entity2]['entity_type'].iloc[0]
    
    emoji1 = entity_emojis.get(type1, '📌')
    emoji2 = entity_emojis.get(type2, '📌')
    
    print(f"{idx:2d}. {emoji1} {entity1:25s} ↔️  {emoji2} {entity2:25s} | {count} kez birlikte")

# Haber kategorilerine göre en çok birlikte geçen entity'ler
print("\n" + "=" * 80)
print("📰 KATEGORİ BAZINDA EN ÇOK BİRLİKTE GEÇEN ENTITY'LER")
print("=" * 80)

for category in news_df['category'].unique():
    category_news_ids = news_df[news_df['category'] == category]['id'].tolist()
    category_entities_df = entities_full_df[entities_full_df['news_id'].isin(category_news_ids)]
    
    # Bu kategorideki entity çiftleri
    category_co_occurrence = defaultdict(int)
    
    for news_id in category_news_ids:
        news_entities = category_entities_df[category_entities_df['news_id'] == news_id]['entity_text'].unique()
        
        for entity1, entity2 in combinations(sorted(news_entities), 2):
            pair = (entity1, entity2)
            category_co_occurrence[pair] += 1
    
    if category_co_occurrence:
        top_pair = max(category_co_occurrence.items(), key=lambda x: x[1])
        entity1, entity2 = top_pair[0]
        count = top_pair[1]
        
        type1 = category_entities_df[category_entities_df['entity_text'] == entity1]['entity_type'].iloc[0]
        type2 = category_entities_df[category_entities_df['entity_text'] == entity2]['entity_type'].iloc[0]
        
        emoji1 = entity_emojis.get(type1, '📌')
        emoji2 = entity_emojis.get(type2, '📌')
        
        print(f"📂 {category:15s} | {emoji1} {entity1} ↔️  {emoji2} {entity2} ({count} kez)")

print("\n✅ Co-occurrence analizi tamamlandı!")
````

### 🕸️ 1. Entity Network Graph (Ağ Analizi)

Entity'lerin aynı haberde geçme ilişkilerini ağ grafiği olarak görselleştireceğiz.

````python
# 🕸️ Entity Network Graph Oluşturma
print("🔄 Entity network graph oluşturuluyor...")

import networkx as nx
from itertools import combinations
from collections import Counter

# Co-occurrence hesapla (aynı haberde geçen entity'ler)
co_occurrence_pairs = []

for news_id in bbc_entities_df['news_id'].unique():
    # Bu haberdeki tüm entity'leri al
    news_entities = bbc_entities_df[bbc_entities_df['news_id'] == news_id]['entity_text'].unique()
    
    # En fazla 10 entity varsa işle (çok fazla kombinasyon olmasın)
    if len(news_entities) > 1 and len(news_entities) <= 10:
        # Tüm ikili kombinasyonları oluştur
        for ent1, ent2 in combinations(news_entities, 2):
            co_occurrence_pairs.append((ent1, ent2))

# En sık birlikte geçen 50 çifti al
pair_counts = Counter(co_occurrence_pairs)
top_pairs = pair_counts.most_common(50)

print(f"✅ Toplam {len(co_occurrence_pairs):,} entity çifti bulundu")
print(f"📊 En sık 50 çift seçildi\n")

# Network graph oluştur
G = nx.Graph()

# Node ve edge ekle
for (ent1, ent2), weight in top_pairs:
    G.add_edge(ent1, ent2, weight=weight)

# Graph istatistikleri
print(f"📊 NETWORK GRAPH İSTATİSTİKLERİ")
print(f"{'='*60}")
print(f"Node sayısı: {G.number_of_nodes()}")
print(f"Edge sayısı: {G.number_of_edges()}")
print(f"Ortalama derece: {sum(dict(G.degree()).values()) / G.number_of_nodes():.2f}")

# En merkezi node'lar (degree centrality)
degree_centrality = nx.degree_centrality(G)
top_central = sorted(degree_centrality.items(), key=lambda x: x[1], reverse=True)[:10]

print(f"\n🎯 EN MERKEZİ ENTITY'LER (Degree Centrality):")
for idx, (entity, centrality) in enumerate(top_central, 1):
    print(f"{idx:2d}. {entity:30s} - {centrality:.3f}")

# Görselleştirme
plt.figure(figsize=(18, 12))

# Layout
pos = nx.spring_layout(G, k=2, iterations=50, seed=42)

# Node size - degree'ye göre
node_sizes = [G.degree(node) * 100 for node in G.nodes()]

# Edge width - weight'e göre
edge_widths = [G[u][v]['weight'] * 0.3 for u, v in G.edges()]

# Çizim
nx.draw_networkx_nodes(G, pos, node_size=node_sizes, 
                       node_color='lightblue', alpha=0.7,
                       edgecolors='darkblue', linewidths=2)

nx.draw_networkx_edges(G, pos, width=edge_widths, 
                       alpha=0.4, edge_color='gray')

nx.draw_networkx_labels(G, pos, font_size=9, font_weight='bold',
                        font_family='sans-serif')

plt.title('🕸️ Entity Network Graph - En Sık Birlikte Geçen Entity\'ler', 
          fontsize=18, fontweight='bold', pad=20)
plt.axis('off')
plt.tight_layout()
plt.show()

print("\n✅ Network graph oluşturuldu!")
````

### 🔗 3. Co-occurrence Matrix (Birlikte Geçme Matrisi)

Hangi entity türlerinin birlikte geçtiğini heatmap ile gösterelim.

````python
# 🔗 Co-occurrence Matrix - Entity Türleri Arası İlişki
print("🔄 Co-occurrence matrix hesaplanıyor...")

from collections import defaultdict
import seaborn as sns

# Entity türleri arası co-occurrence hesapla
entity_type_cooccurrence = defaultdict(lambda: defaultdict(int))

for news_id in bbc_entities_df['news_id'].unique()[:1000]:  # İlk 1000 haber
    # Bu haberdeki entity türlerini al
    news_entity_types = bbc_entities_df[bbc_entities_df['news_id'] == news_id]['entity_type'].tolist()
    
    # Birlikte geçen türleri say
    for i, type1 in enumerate(news_entity_types):
        for type2 in news_entity_types[i+1:]:
            if type1 != type2:
                entity_type_cooccurrence[type1][type2] += 1
                entity_type_cooccurrence[type2][type1] += 1

# En sık görülen entity türlerini al
top_entity_types = bbc_entities_df['entity_type'].value_counts().head(10).index.tolist()

# Matrix oluştur
cooccurrence_matrix = pd.DataFrame(0, index=top_entity_types, columns=top_entity_types)

for type1 in top_entity_types:
    for type2 in top_entity_types:
        if type1 in entity_type_cooccurrence and type2 in entity_type_cooccurrence[type1]:
            cooccurrence_matrix.loc[type1, type2] = entity_type_cooccurrence[type1][type2]

print(f"✅ Co-occurrence matrix hesaplandı!\n")

# Görselleştirme - Matplotlib ile
fig, ax = plt.subplots(figsize=(14, 12))

sns.heatmap(
    cooccurrence_matrix, 
    annot=True, 
    fmt='g', 
    cmap='YlOrRd',
    square=True,
    linewidths=1,
    cbar_kws={'label': 'Birlikte Geçme Sayısı'},
    ax=ax
)

plt.title('🔗 Entity Türleri Co-occurrence Matrix\n(Aynı Haberde Birlikte Geçme Sıklığı)', 
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Entity Türü', fontsize=13, fontweight='bold')
plt.ylabel('Entity Türü', fontsize=13, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# İnteraktif Plotly Heatmap
fig_interactive = go.Figure(data=go.Heatmap(
    z=cooccurrence_matrix.values,
    x=cooccurrence_matrix.columns,
    y=cooccurrence_matrix.index,
    colorscale='YlOrRd',
    text=cooccurrence_matrix.values,
    texttemplate='%{text}',
    textfont={"size": 11},
    hovertemplate='%{y} ↔️ %{x}<br>Birlikte: %{z} kez<extra></extra>',
    colorbar=dict(title='Birlikte Geçme')
))

fig_interactive.update_layout(
    title='🔗 Entity Türleri Co-occurrence Matrix (İnteraktif)',
    title_font_size=18,
    xaxis_title='Entity Türü',
    yaxis_title='Entity Türü',
    height=700,
    width=800
)

fig_interactive.show()

# En çok birlikte geçen türleri listele
print("\n📊 EN ÇOK BİRLİKTE GEÇEN ENTITY TÜRLERİ")
print("="*60)

cooccurrence_list = []
for type1 in top_entity_types:
    for type2 in top_entity_types:
        if type1 < type2:  # Tekrarları önle
            count = cooccurrence_matrix.loc[type1, type2]
            if count > 0:
                cooccurrence_list.append((type1, type2, count))

cooccurrence_list.sort(key=lambda x: x[2], reverse=True)

for idx, (type1, type2, count) in enumerate(cooccurrence_list[:15], 1):
    print(f"{idx:2d}. {type1:12s} ↔️ {type2:12s}: {count:4d} kez")

print("\n✅ Co-occurrence analizi tamamlandı!")
````

### ⚡ Model Performans Karşılaştırması

Aynı haber metinlerini farklı modellerle işleyip hız ve entity sayısını karşılaştıralım.

````python
# ⚡ Model Performans Karşılaştırması
print("🔄 Model performansları test ediliyor...\n")

# Test için örnek metinler (ilk 100 haber)
test_texts = bbc_df['text'].head(100).tolist()

# Sonuçları saklayacak dictionary
benchmark_results = {}

# Her modeli test et
for model_name, model_info in available_models.items():
    print(f"🧪 Test ediliyor: {model_name.upper()} ({model_info['name']})")
    
    model = model_info['model']
    
    start_time = time.time()
    total_entities = 0
    entity_types_found = set()
    
    for text in test_texts:
        doc = model(text[:10000])  # İlk 10k karakter
        total_entities += len(doc.ents)
        entity_types_found.update([ent.label_ for ent in doc.ents])
    
    elapsed_time = time.time() - start_time
    
    benchmark_results[model_name] = {
        'model_name': model_info['name'],
        'total_time': elapsed_time,
        'avg_time_per_doc': elapsed_time / len(test_texts),
        'total_entities': total_entities,
        'avg_entities_per_doc': total_entities / len(test_texts),
        'unique_entity_types': len(entity_types_found),
        'docs_per_second': len(test_texts) / elapsed_time
    }
    
    print(f"   ⏱️ Süre: {elapsed_time:.2f}s | 📊 Entity: {total_entities} | "
          f"🚀 Hız: {len(test_texts) / elapsed_time:.1f} dok/sn")
    print()

# Sonuçları DataFrame'e dönüştür
benchmark_df = pd.DataFrame(benchmark_results).T

print("\n" + "="*80)
print("📊 MODEL PERFORMANS KARŞILAŞTIRMASI")
print("="*80)
print(f"\nTest Edilen Doküman Sayısı: {len(test_texts)}\n")

# Tablo formatında göster
print(f"{'Model':<15} {'Toplam Süre':<12} {'Ortalama Süre':<15} {'Toplam Entity':<15} {'Hız (dok/sn)':<12}")
print("-" * 80)

for model_name, results in benchmark_results.items():
    print(f"{model_name.upper():<15} "
          f"{results['total_time']:>10.2f}s  "
          f"{results['avg_time_per_doc']:>12.4f}s  "
          f"{results['total_entities']:>13d}  "
          f"{results['docs_per_second']:>10.1f}")

# Görselleştirme 1: İşlem Hızı Karşılaştırması
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Toplam İşlem Süresi
ax1 = axes[0, 0]
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4']
bars1 = ax1.bar(benchmark_df.index, benchmark_df['total_time'], color=colors[:len(benchmark_df)])
ax1.set_title('⏱️ Toplam İşlem Süresi', fontsize=14, fontweight='bold')
ax1.set_ylabel('Süre (saniye)', fontsize=11)
ax1.set_xlabel('Model', fontsize=11)
for bar in bars1:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.2f}s', ha='center', va='bottom', fontweight='bold')

# 2. Doküman Başına Süre
ax2 = axes[0, 1]
bars2 = ax2.bar(benchmark_df.index, benchmark_df['avg_time_per_doc'] * 1000, color=colors[:len(benchmark_df)])
ax2.set_title('⚡ Doküman Başına İşlem Süresi', fontsize=14, fontweight='bold')
ax2.set_ylabel('Süre (milisaniye)', fontsize=11)
ax2.set_xlabel('Model', fontsize=11)
for bar in bars2:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.1f}ms', ha='center', va='bottom', fontweight='bold')

# 3. Bulunan Entity Sayısı
ax3 = axes[1, 0]
bars3 = ax3.bar(benchmark_df.index, benchmark_df['total_entities'], color=colors[:len(benchmark_df)])
ax3.set_title('📊 Bulunan Toplam Entity Sayısı', fontsize=14, fontweight='bold')
ax3.set_ylabel('Entity Sayısı', fontsize=11)
ax3.set_xlabel('Model', fontsize=11)
for bar in bars3:
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2., height,
            f'{int(height)}', ha='center', va='bottom', fontweight='bold')

# 4. İşlem Hızı (dok/saniye)
ax4 = axes[1, 1]
bars4 = ax4.bar(benchmark_df.index, benchmark_df['docs_per_second'], color=colors[:len(benchmark_df)])
ax4.set_title('🚀 İşlem Hızı (Doküman/Saniye)', fontsize=14, fontweight='bold')
ax4.set_ylabel('Doküman/Saniye', fontsize=11)
ax4.set_xlabel('Model', fontsize=11)
for bar in bars4:
    height = bar.get_height()
    ax4.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.1f}', ha='center', va='bottom', fontweight='bold')

plt.suptitle('🤖 spaCy Model Performans Karşılaştırması', 
             fontsize=18, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

# En iyi performans
fastest_model = benchmark_df['docs_per_second'].idxmax()
most_entities = benchmark_df['total_entities'].idxmax()

print(f"\n🏆 PERFORMANS ÖZETİ")
print(f"{'='*60}")
print(f"⚡ En hızlı model: {fastest_model.upper()} "
      f"({benchmark_df.loc[fastest_model, 'docs_per_second']:.1f} dok/sn)")
print(f"📊 En fazla entity bulan: {most_entities.upper()} "
      f"({int(benchmark_df.loc[most_entities, 'total_entities'])} entity)")

print("\n✅ Model karşılaştırması tamamlandı!")
````
